```python
# ==========================================================
# COMPLETE CODE FROM START TO END
# SINGLE TASK: DEPRESSION LEVELS ONLY (4 CLASSES)
# MRI FEATURES ALREADY EXTRACTED
# LATE FUSION AFTER TRANSFORMER-BIGRU (PARALLEL BRANCHES)
#   -> Transformer and BiGRU run IN PARALLEL on clinical data
#   -> MRI/image features fused at the END via 3 vectors:
#        tr_ctx (Transformer attention context)
#        g_last (last BiGRU hidden state)
#        mri_ctx (MRI branch embedding)
# MULTI-SEED (5 seeds) - NO FOLD (patient-wise 70/15/15 split)
# T2 / T4 / T6 + OPTUNA
# CPU VERSION
# ==========================================================

import re
import json
import random
import gc
import warnings
from pathlib import Path
from functools import reduce
from contextlib import nullcontext

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.model_selection import GroupShuffleSplit
from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import (
    classification_report,
    confusion_matrix,
    f1_score,
    balanced_accuracy_score,
    accuracy_score,
    precision_score,
    recall_score,
)

import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import DataLoader, TensorDataset, WeightedRandomSampler

import optuna
from optuna.samplers import TPESampler
from optuna.pruners import MedianPruner

warnings.filterwarnings("ignore")

try:
    from IPython.display import display
except Exception:
    def display(x):
        print(x)


# ==========================================================
# 0) CONFIG
# ==========================================================
OUT_DIR = Path("/kaggle/working/levels_late_fusion_parallel_multi_seed_no_fold")
OUT_DIR.mkdir(parents=True, exist_ok=True)

HORIZONS = {
    "T2": {"months": [0, 12], "last_month": 12},
    "T4": {"months": [0, 12, 24, 36], "last_month": 36},
    "T6": {"months": [0, 12, 24, 36, 48, 60], "last_month": 60},
}

# ✅ Multi-seed, sans fold
SEEDS = [42, 43, 44, 45, 46]
BASE_SEED = 42

TRAIN_RATIO = 0.70
VAL_RATIO = 0.15
TEST_RATIO = 0.15
assert abs(TRAIN_RATIO + VAL_RATIO + TEST_RATIO - 1.0) < 1e-9

N_TRIALS = 20
GRAD_CLIP = 1.0
MRI_MISSING_FILL = -1.0
NUM_CLASSES = 4

CLASS_NAMES = [
    "Level0_Minimal(0-4)",
    "Level1_Mild(5-8)",
    "Level2_Moderate(9-11)",
    "Level3_Severe(12+)",
]

NUM_WORKERS = 0
PIN_MEMORY = False

device = torch.device("cpu")
print("device:", device)


# ==========================================================
# 1) SEED
# ==========================================================
def set_seed(seed=42):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    if torch.cuda.is_available():
        torch.cuda.manual_seed_all(seed)
    if hasattr(torch.backends, "cudnn"):
        torch.backends.cudnn.deterministic = True
        torch.backends.cudnn.benchmark = False


set_seed(BASE_SEED)


def autocast_context():
    return nullcontext()


# ==========================================================
# 2) FILE HELPERS
# ==========================================================
def find_file(filename):
    base = Path("/kaggle/input")
    matches = list(base.rglob(filename))
    if not matches:
        raise FileNotFoundError(f"{filename} introuvable dans /kaggle/input")
    print(f"[FOUND] {filename} -> {matches[0]}")
    return matches[0]


def read_csv_robust(path):
    return pd.read_csv(
        path,
        encoding="latin1",
        sep=None,
        engine="python",
        on_bad_lines="skip"
    )


# ==========================================================
# 3) ROBUST MRI FILE DETECTION (identique à ton code)
# ==========================================================
def list_possible_mri_files():
    base = Path("/kaggle/input")
    files = []
    for ext in ["*.csv", "*.xlsx", "*.xls", "*.parquet"]:
        files.extend(list(base.rglob(ext)))

    print("===== POSSIBLE FILES IN /kaggle/input =====")
    for f in files:
        print(f)
    print("===========================================")
    return files


def read_table_flexible(fp, nrows=None):
    fp = Path(fp)
    suffix = fp.suffix.lower()

    if suffix == ".csv":
        if nrows is None:
            return pd.read_csv(fp)
        return pd.read_csv(fp, nrows=nrows)

    elif suffix in [".xlsx", ".xls"]:
        if nrows is None:
            return pd.read_excel(fp)
        return pd.read_excel(fp, nrows=nrows)

    elif suffix == ".parquet":
        df = pd.read_parquet(fp)
        if nrows is not None:
            return df.head(nrows)
        return df

    else:
        raise ValueError(f"Unsupported file type: {fp}")


def detect_patno_column(cols):
    cols_lower = [str(c).strip().lower() for c in cols]

    patno_candidates = [
        "patno", "patient_id", "patientid", "subject_id", "subjectid", "id"
    ]

    for c in cols_lower:
        if c in patno_candidates:
            return c

    for c in cols_lower:
        if "patno" in c or "patient" in c or "subject" in c:
            return c

    return None


def detect_feature_columns(cols):
    cols_lower = [str(c).strip().lower() for c in cols]

    feature_cols = []

    for c in cols_lower:
        if c.startswith("mri_feat") or c.startswith("feat") or c.startswith("feature"):
            feature_cols.append(c)

    if len(feature_cols) == 0:
        excluded_keywords = [
            "patno", "patient", "subject", "id",
            "available", "flag", "label", "target",
            "event", "month", "time", "visit"
        ]
        for c in cols_lower:
            if not any(k in c for k in excluded_keywords):
                feature_cols.append(c)

    return feature_cols


def find_mri_feature_file():
    files = list_possible_mri_files()
    valid_files = []

    for fp in files:
        try:
            df_head = read_table_flexible(fp, nrows=5)
            cols_original = [str(c).strip() for c in df_head.columns]

            patno_col = detect_patno_column(cols_original)
            feat_cols = detect_feature_columns(cols_original)

            if patno_col is not None and len(feat_cols) >= 32:
                valid_files.append((fp, patno_col, feat_cols, cols_original))
        except Exception:
            pass

    if len(valid_files) == 0:
        raise FileNotFoundError(
            "Aucun fichier MRI valide trouvé.\n"
            "Il faut un fichier avec une colonne patient (PATNO ou similaire)\n"
            "et beaucoup de colonnes features."
        )

    print("\n===== MRI CANDIDATES FOUND =====")
    for i, (fp, patno_col, feat_cols, cols_original) in enumerate(valid_files):
        print(f"[{i}] File: {fp}")
        print(f"    PATNO column: {patno_col}")
        print(f"    Number of feature columns: {len(feat_cols)}")
        print(f"    First 10 columns: {cols_original[:10]}")
    print("================================\n")

    chosen_fp, chosen_patno, chosen_feat_cols, _ = valid_files[0]
    print("[SELECTED MRI FILE] ->", chosen_fp)
    print("[PATNO COLUMN] ->", chosen_patno)
    print("[NB FEATURES] ->", len(chosen_feat_cols))

    return chosen_fp, chosen_patno, chosen_feat_cols


def load_preextracted_mri_features_auto():
    mri_fp, patno_col_detected, feat_cols_detected = find_mri_feature_file()

    mri_df = read_table_flexible(mri_fp, nrows=None)
    original_cols = [str(c).strip() for c in mri_df.columns]
    lower_to_original = {str(c).strip().lower(): str(c).strip() for c in original_cols}

    patno_col_real = lower_to_original[patno_col_detected]
    feat_cols_real = [lower_to_original[c] for c in feat_cols_detected if c in lower_to_original]

    mri_df = mri_df.copy()
    mri_df[patno_col_real] = pd.to_numeric(mri_df[patno_col_real], errors="coerce")
    mri_df = mri_df.dropna(subset=[patno_col_real]).copy()
    mri_df[patno_col_real] = mri_df[patno_col_real].astype(int)

    available_col = None
    for c in original_cols:
        cl = c.lower()
        if "available" in cl or "flag" in cl:
            available_col = c
            break

    if available_col is None:
        mri_df["__MRI_AVAILABLE__"] = 1
        available_col = "__MRI_AVAILABLE__"

    good_feat_cols = []
    for c in feat_cols_real:
        vals = pd.to_numeric(mri_df[c], errors="coerce")
        if vals.notna().sum() > 0:
            mri_df[c] = vals
            good_feat_cols.append(c)

    if len(good_feat_cols) == 0:
        raise ValueError("No usable MRI feature columns found after numeric conversion.")

    feat_dim = len(good_feat_cols)
    print("[FINAL MRI FEATURE DIM] ->", feat_dim)

    feat_dict = {}
    for _, row in mri_df.iterrows():
        patno = int(row[patno_col_real])
        available = float(row[available_col]) if pd.notna(row[available_col]) else 0.0

        if available > 0.5:
            feat = row[good_feat_cols].to_numpy(dtype=np.float32)
            feat = np.nan_to_num(feat, nan=0.0, posinf=0.0, neginf=0.0)
            feat_dict[patno] = {
                "feat": feat,
                "available": 1.0
            }
        else:
            feat_dict[patno] = {
                "feat": np.full(feat_dim, MRI_MISSING_FILL, dtype=np.float32),
                "available": 0.0
            }

    return feat_dict, good_feat_cols, feat_dim

# ==========================================================
# 4) MERGE HELPERS
# ==========================================================
KEYS = ["PATNO", "EVENT_ID"]
DATE_CANDIDATES = ["INFODT", "VISITDT", "DATE", "EXAMDATE", "TESTDATE"]

ADMIN_WORDS = [
    "REC_ID", "PAG_NAME", "ORIG_ENTRY", "LAST_UPDATE",
    "ENTRY", "UPDATE", "NUPSOURC", "PTCGBOTH",
    "INFODT", "VISITDT", "EXAMDATE", "TESTDATE", "DATE"
]

LEAKY_CONTAINS = [
    "ESS_TOTAL", "SCORE_GDS", "SCAUTOT", "RBDTOT",
    "NP1RTOT", "NP2PTOT", "NP3TOT", "NHY"
]

_FR_MONTH = {
    "janv": "Jan", "févr": "Feb", "fevr": "Feb", "mars": "Mar", "avr": "Apr",
    "mai": "May", "juin": "Jun", "juil": "Jul", "août": "Aug", "aout": "Aug",
    "sept": "Sep", "oct": "Oct", "nov": "Nov", "déc": "Dec", "dec": "Dec"
}


def norm_cols(df):
    df = df.copy()
    df.columns = [str(c).strip().upper() for c in df.columns]
    return df


def coerce_keys(df):
    df = df.copy()
    if "PATNO" in df.columns:
        df["PATNO"] = pd.to_numeric(df["PATNO"], errors="coerce").astype("Int64")
    if "EVENT_ID" in df.columns:
        df["EVENT_ID"] = df["EVENT_ID"].astype(str).str.upper().str.strip()
    return df


def detect_date_col(df):
    for c in DATE_CANDIDATES:
        if c in df.columns:
            return c
    return None


def parse_date_series(s):
    s = s.astype(str).str.strip()
    for fr, en in _FR_MONTH.items():
        s = s.str.replace(fr, en, regex=False)
    return pd.to_datetime(s, errors="coerce", dayfirst=True)


def dedup_by_key_keep_latest(df, keys=None):
    df = df.copy()

    if keys is None:
        keys = [k for k in ["PATNO", "EVENT_ID"] if k in df.columns]

    if len(keys) == 0:
        return df.drop_duplicates(keep="last")

    dcol = detect_date_col(df)

    if dcol is not None:
        df["_DATE_TMP_"] = parse_date_series(df[dcol])
        df = df.sort_values(keys + ["_DATE_TMP_"])
        df = df.drop_duplicates(keys, keep="last")
        df = df.drop(columns=["_DATE_TMP_"])
    else:
        df = df.drop_duplicates(keys, keep="last")

    return df


def prefix_features(df, prefix, protected=("PATNO", "EVENT_ID")):
    df = df.copy()
    new_cols = {}
    for c in df.columns:
        if c not in protected:
            new_cols[c] = f"{prefix}{c}"
    return df.rename(columns=new_cols)


def event_id_to_month(x):
    x = str(x).strip().upper()
    if x == "BL":
        return 0
    m = re.fullmatch(r"V(\d+)", x)
    if m:
        v = int(m.group(1))
        return (v - 2) * 6
    return np.nan


def drop_admin_cols_safe(df):
    cols = []
    for c in df.columns:
        cu = c.upper()
        if cu in KEYS:
            continue
        if any(w in cu for w in ADMIN_WORDS):
            cols.append(c)
    return df.drop(columns=cols, errors="ignore")


def drop_leaky_cols_safe(df):
    cols = []
    for c in df.columns:
        cu = c.upper()
        if cu in KEYS:
            continue
        if any(w in cu for w in LEAKY_CONTAINS):
            cols.append(c)
    return df.drop(columns=cols, errors="ignore")


def filter_to_pd(participant_status):
    df = participant_status.copy()
    candidate_cols = [c for c in df.columns if "COHORT" in c or "STATUS" in c or "DEF" in c]

    if len(candidate_cols) == 0:
        txt = df.astype(str).agg(" ".join, axis=1)
    else:
        txt = pd.Series("", index=df.index)
        for c in candidate_cols:
            txt = txt.astype(str) + " " + df[c].astype(str)

    mask = txt.str.contains("PARKINSON", case=False, na=False)
    keep_pat = df.loc[mask, "PATNO"].dropna().astype(int).unique()
    return set(keep_pat)


def drop_low_variance_columns(df, cols, threshold=1e-8):
    keep = []
    for c in cols:
        vals = pd.to_numeric(df[c], errors="coerce")
        if vals.notna().sum() == 0:
            continue
        if vals.var(skipna=True) > threshold:
            keep.append(c)
    return keep


# ==========================================================
# 5) LOAD CLINICAL FILES
# ==========================================================
demo_path   = find_file("Demographics.csv")
ess_path    = find_file("Epworth_Sleepiness_Scale.csv")
up1_path    = find_file("MDS-UPDRS_Part_I.csv")
up4_path    = find_file("MDS-UPDRS_Part_IV.csv")
rem_path    = find_file("REM_Sleep_Behavior.csv")
scopa_path  = find_file("SCOPA-AUT.csv")
gds_path    = find_file("Geriatric_Depression_Scale.csv")
status_path = find_file("Participant_Status.csv")
up23_path   = find_file("UPDRS_PartII_PartIII.xlsx")

demo = prefix_features(
    dedup_by_key_keep_latest(coerce_keys(norm_cols(read_csv_robust(demo_path))), keys=["PATNO", "EVENT_ID"]),
    "DEMO__"
)
ess = prefix_features(
    dedup_by_key_keep_latest(coerce_keys(norm_cols(read_csv_robust(ess_path))), keys=["PATNO", "EVENT_ID"]),
    "ESS__"
)
up1 = prefix_features(
    dedup_by_key_keep_latest(coerce_keys(norm_cols(read_csv_robust(up1_path))), keys=["PATNO", "EVENT_ID"]),
    "UP1__"
)
up4 = prefix_features(
    dedup_by_key_keep_latest(coerce_keys(norm_cols(read_csv_robust(up4_path))), keys=["PATNO", "EVENT_ID"]),
    "UP4__"
)
rem = prefix_features(
    dedup_by_key_keep_latest(coerce_keys(norm_cols(read_csv_robust(rem_path))), keys=["PATNO", "EVENT_ID"]),
    "REM__"
)
scopa = prefix_features(
    dedup_by_key_keep_latest(coerce_keys(norm_cols(read_csv_robust(scopa_path))), keys=["PATNO", "EVENT_ID"]),
    "SCOPA__"
)
gds = prefix_features(
    dedup_by_key_keep_latest(coerce_keys(norm_cols(read_csv_robust(gds_path))), keys=["PATNO", "EVENT_ID"]),
    "GDS__"
)
status = dedup_by_key_keep_latest(coerce_keys(norm_cols(read_csv_robust(status_path))), keys=["PATNO"])

up23 = pd.read_excel(up23_path)
up23 = prefix_features(
    dedup_by_key_keep_latest(coerce_keys(norm_cols(up23)), keys=["PATNO", "EVENT_ID"]),
    "UP23__"
)

demo  = drop_leaky_cols_safe(drop_admin_cols_safe(demo))
ess   = drop_leaky_cols_safe(drop_admin_cols_safe(ess))
up1   = drop_leaky_cols_safe(drop_admin_cols_safe(up1))
up4   = drop_leaky_cols_safe(drop_admin_cols_safe(up4))
rem   = drop_leaky_cols_safe(drop_admin_cols_safe(rem))
scopa = drop_leaky_cols_safe(drop_admin_cols_safe(scopa))
gds   = drop_leaky_cols_safe(drop_admin_cols_safe(gds))
up23  = drop_leaky_cols_safe(drop_admin_cols_safe(up23))

dfs = [demo, ess, up1, up4, rem, scopa, gds, up23]
df_seq = reduce(lambda left, right: pd.merge(left, right, on=KEYS, how="outer"), dfs)

status = coerce_keys(norm_cols(status))
pd_patients = filter_to_pd(status)
df_seq = df_seq[df_seq["PATNO"].isin(pd_patients)].copy()

df_seq["MONTH"] = df_seq["EVENT_ID"].apply(event_id_to_month)
df_seq = df_seq[df_seq["MONTH"].notna()].copy()
df_seq["MONTH"] = df_seq["MONTH"].astype(int)

df_seq = df_seq.sort_values(["PATNO", "MONTH"])
df_seq = df_seq.drop_duplicates(subset=["PATNO", "MONTH"], keep="last").copy()

print("Merged df_seq shape:", df_seq.shape)


# ==========================================================
# 6) BUILD LEVELS LABEL ONLY (4 classes)
# ==========================================================
if "GDS__LEVELS" not in df_seq.columns:
    gds_score_candidates = [c for c in df_seq.columns if "GDS" in c and ("TOTAL" in c or "SCORE" in c)]
    if len(gds_score_candidates) == 0:
        raise ValueError("GDS__LEVELS not found and no GDS score column detected.")
    gds_score_col = gds_score_candidates[0]
    score = pd.to_numeric(df_seq[gds_score_col], errors="coerce")

    levels = pd.Series(np.nan, index=df_seq.index)
    levels[(score >= 0) & (score <= 4)] = 0
    levels[(score >= 5) & (score <= 8)] = 1
    levels[(score >= 9) & (score <= 11)] = 2
    levels[(score >= 12)] = 3
    df_seq["GDS__LEVELS"] = levels.astype("Float64")

print("Levels labels ready.")


# ==========================================================
# 7) KEEP NUMERIC CLINICAL FEATURES
# ==========================================================
protected_cols = ["PATNO", "EVENT_ID", "MONTH", "GDS__LEVELS"]

for c in df_seq.columns:
    if c not in ["PATNO", "EVENT_ID"]:
        df_seq[c] = pd.to_numeric(df_seq[c], errors="ignore")

numeric_cols = []
for c in df_seq.columns:
    if c in protected_cols:
        continue
    if pd.api.types.is_numeric_dtype(df_seq[c]):
        numeric_cols.append(c)

feature_cols_model = []
for c in numeric_cols:
    miss = df_seq[c].isna().mean()
    protected_prefix = c.startswith(("ESS__", "REM__", "SCOPA__", "GDS__", "UP1__", "UP4__", "UP23__", "DEMO__"))
    if miss <= 0.90 or protected_prefix:
        feature_cols_model.append(c)

feature_cols_model = drop_low_variance_columns(df_seq, feature_cols_model)
print("Clinical feature columns:", len(feature_cols_model))

CLINICAL_RAW_DIM = len(feature_cols_model)


# ==========================================================
# 8) LOAD MRI FEATURES AUTO
# ==========================================================
mri_feat_dict, MRI_FEATURE_COLUMNS, MRI_FEAT_DIM = load_preextracted_mri_features_auto()
FLAG_DIM = 1
print("✅ Clinical raw dim:", CLINICAL_RAW_DIM)
print("✅ MRI feature dim:", MRI_FEAT_DIM)

# ==========================================================
# 9) MAKE SEQUENCES - RAW FUSED LAYOUT
#    [clinical_raw | mri_raw | mri_flag] per time step.
#    The MODEL itself splits this layout and processes:
#      - clinical -> Transformer (parallel) + BiGRU (parallel)
#      - MRI (last time point) -> separate MRI branch
#    then fuses tr_ctx + g_last + mri_ctx at the very end.
# ==========================================================
def make_levels_sequences_raw_fused(df_seq, feature_cols, months, last_month):
    Xfused_list, pat_list = [], []
    y_list = []

    for patno, g in df_seq.groupby("PATNO"):
        g = g.copy()
        g = g.sort_values("MONTH")
        g = g.drop_duplicates(subset=["MONTH"], keep="last")
        g = g.set_index("MONTH")

        if last_month not in g.index:
            continue

        gg = g.reindex(months)
        Xc = gg[feature_cols].to_numpy(dtype=np.float32)

        row_last = g.loc[last_month]
        if isinstance(row_last, pd.DataFrame):
            row_last = row_last.iloc[-1]

        y_lvl = row_last["GDS__LEVELS"] if "GDS__LEVELS" in row_last.index else pd.NA
        if pd.isna(y_lvl):
            continue

        if int(patno) in mri_feat_dict:
            xmri = mri_feat_dict[int(patno)]["feat"].astype(np.float32)
            xflag = np.array([mri_feat_dict[int(patno)]["available"]], dtype=np.float32)
        else:
            xmri = np.full(MRI_FEAT_DIM, MRI_MISSING_FILL, dtype=np.float32)
            xflag = np.array([0.0], dtype=np.float32)

        T = Xc.shape[0]
        xmri_rep = np.repeat(xmri.reshape(1, -1), T, axis=0)
        xflag_rep = np.repeat(xflag.reshape(1, -1), T, axis=0)

        Xfused = np.concatenate([Xc, xmri_rep, xflag_rep], axis=1).astype(np.float32)

        Xfused_list.append(Xfused)
        pat_list.append(int(patno))
        y_list.append(int(y_lvl))

    if len(Xfused_list) == 0:
        raise ValueError("Aucune séquence construite.")

    return (
        np.stack(Xfused_list).astype(np.float32),
        np.array(y_list, dtype=np.int64),
        np.array(pat_list, dtype=np.int64),
    )


X2, y2, p2 = make_levels_sequences_raw_fused(df_seq, feature_cols_model, HORIZONS["T2"]["months"], HORIZONS["T2"]["last_month"])
X4, y4, p4 = make_levels_sequences_raw_fused(df_seq, feature_cols_model, HORIZONS["T4"]["months"], HORIZONS["T4"]["last_month"])
X6, y6, p6 = make_levels_sequences_raw_fused(df_seq, feature_cols_model, HORIZONS["T6"]["months"], HORIZONS["T6"]["last_month"])

print("✅ T2 fused:", X2.shape, "unique patients:", len(np.unique(p2)))
print("✅ T4 fused:", X4.shape, "unique patients:", len(np.unique(p4)))
print("✅ T6 fused:", X6.shape, "unique patients:", len(np.unique(p6)))


# ==========================================================
# 10) SPLIT PATIENT-WISE 70/15/15 (SANS FOLD)
# ==========================================================
def split_train_val_test_patientwise(X, y, pat, train_ratio=0.70, val_ratio=0.15, test_ratio=0.15, seed=42):
    assert abs(train_ratio + val_ratio + test_ratio - 1.0) < 1e-9

    unique_pat = np.unique(pat)

    gss1 = GroupShuffleSplit(n_splits=1, test_size=(val_ratio + test_ratio), random_state=seed)
    tr_pat_idx, tmp_pat_idx = next(gss1.split(unique_pat, groups=unique_pat))
    pat_tr_set = set(unique_pat[tr_pat_idx])
    pat_tmp = unique_pat[tmp_pat_idx]

    test_share_of_temp = test_ratio / (val_ratio + test_ratio)
    gss2 = GroupShuffleSplit(n_splits=1, test_size=test_share_of_temp, random_state=seed)
    va_pat_idx, te_pat_idx = next(gss2.split(pat_tmp, groups=pat_tmp))
    pat_va_set = set(pat_tmp[va_pat_idx])
    pat_te_set = set(pat_tmp[te_pat_idx])

    tr_idx = np.array([i for i, pp in enumerate(pat) if pp in pat_tr_set], dtype=int)
    va_idx = np.array([i for i, pp in enumerate(pat) if pp in pat_va_set], dtype=int)
    te_idx = np.array([i for i, pp in enumerate(pat) if pp in pat_te_set], dtype=int)

    return (
        (X[tr_idx], y[tr_idx], pat[tr_idx]),
        (X[va_idx], y[va_idx], pat[va_idx]),
        (X[te_idx], y[te_idx], pat[te_idx]),
    )


def print_split_info(name, tr, va, te):
    X_tr, y_tr, pat_tr = tr
    X_va, y_va, pat_va = va
    X_te, y_te, pat_te = te
    print(f"\n{name} SPLIT:")
    print("Train:", X_tr.shape, "patients:", len(np.unique(pat_tr)), "class:", np.unique(y_tr, return_counts=True))
    print("Val  :", X_va.shape, "patients:", len(np.unique(pat_va)), "class:", np.unique(y_va, return_counts=True))
    print("Test :", X_te.shape, "patients:", len(np.unique(pat_te)), "class:", np.unique(y_te, return_counts=True))


# ==========================================================
# 11) PREPROCESSING TRAIN ONLY + NaN MASK
#     Output layout: [scaled(clinical|mri|flag) | mask(clinical|mri|flag)]
#     -> matches exactly what the model's split function expects.
# ==========================================================
def fit_preprocess_train_only_with_mask(X_train):
    X_train = X_train.astype(np.float32)
    N, T, Fdim = X_train.shape

    mask = np.isfinite(X_train).astype(np.float32)

    flat = X_train.reshape(-1, Fdim)
    med = np.nanmedian(flat, axis=0)
    med = np.where(np.isfinite(med), med, 0.0)

    flat_imp = flat.copy()
    inds = np.where(~np.isfinite(flat_imp))
    flat_imp[inds] = np.take(med, inds[1])

    scaler = MinMaxScaler()
    flat_scaled = scaler.fit_transform(flat_imp)
    X_scaled = flat_scaled.reshape(N, T, Fdim).astype(np.float32)

    X_out = np.concatenate([X_scaled, mask], axis=-1).astype(np.float32)
    return X_out, med.astype(np.float32), scaler


def transform_with_train_params_with_mask(X, med, scaler):
    X = X.astype(np.float32)
    N, T, Fdim = X.shape

    mask = np.isfinite(X).astype(np.float32)

    flat = X.reshape(-1, Fdim)
    flat_imp = flat.copy()
    inds = np.where(~np.isfinite(flat_imp))
    flat_imp[inds] = np.take(med, inds[1])

    flat_scaled = scaler.transform(flat_imp)
    X_scaled = flat_scaled.reshape(N, T, Fdim).astype(np.float32)

    X_out = np.concatenate([X_scaled, mask], axis=-1).astype(np.float32)
    return X_out

# ==========================================================
# 12) LOSS (multiclass focal, avec poids inverse-fréquence)
# ==========================================================
def inverse_freq_weights(y, n_classes, power=1.0, max_w=8.0):
    y = np.asarray(y, dtype=np.int64)
    counts = np.bincount(y, minlength=n_classes).astype(np.float32)
    counts[counts == 0] = 1.0
    w = (counts.sum() / counts) ** power
    w = w / w.mean()
    w = np.clip(w, 1.0, max_w)
    return torch.tensor(w, dtype=torch.float32, device=device)


class FocalLossMulti(nn.Module):
    def __init__(self, gamma=2.0, weight=None, reduction="mean"):
        super().__init__()
        self.gamma = gamma
        self.weight = weight
        self.reduction = reduction

    def forward(self, logits, targets):
        ce = F.cross_entropy(logits, targets, weight=self.weight, reduction="none")
        pt = torch.exp(-ce)
        loss = ((1 - pt) ** self.gamma) * ce
        if self.reduction == "mean":
            return loss.mean()
        return loss.sum()


# ==========================================================
# 13) AUGMENTATION (numpy, appliquée une fois avant l'entraînement)
# ==========================================================
def augment_noise(X, scale=0.01):
    if scale <= 0:
        return X.astype(np.float32)
    noise = np.random.normal(0, scale, X.shape).astype(np.float32)
    return (X + noise).astype(np.float32)


def apply_input_dropout(X, p=0.05, fill_value=0.0):
    if p is None or p <= 0:
        return X.astype(np.float32)
    X = X.astype(np.float32, copy=True)
    mask = (np.random.rand(*X.shape) < p)
    X[mask] = np.float32(fill_value)
    return X


def make_sampler(y, n_classes):
    counts = np.bincount(y, minlength=n_classes).astype(np.float32)
    counts[counts == 0] = 1.0
    inv = 1.0 / counts
    sample_w = inv[np.clip(y, 0, n_classes - 1)]
    sample_w = torch.tensor(sample_w, dtype=torch.float32)
    return WeightedRandomSampler(sample_w, num_samples=len(sample_w), replacement=True)


# ==========================================================
# 14) MODEL: TRANSFORMER + BiGRU EN PARALLELE
#     LATE FUSION A 3 BRANCHES (tr_ctx, g_last, mri_ctx)
# ==========================================================
class LevelsParallelLateFusionModel(nn.Module):
    """
    Late fusion à 3 branches :
      1) Transformer (attention pooling)  -> tr_ctx
      2) BiGRU (dernier état caché)       -> g_last
         (Transformer et BiGRU reçoivent TOUS LES DEUX
          directement les données cliniques, EN PARALLELE,
          pas en série comme un Transformer->GRU classique)
      3) Branche MRI/image (dernier point temporel) -> mri_ctx
    Fusion finale : concat([tr_ctx, g_last, mri_ctx]) -> tête de classification (4 classes)

    Input layout après preprocessing :
      [scaled clinical | scaled MRI | scaled MRI_flag | mask clinical | mask MRI | mask MRI_flag]
    """
    def __init__(
        self,
        clinical_raw_dim,
        mri_feat_dim,
        flag_dim=1,
        max_len=6,
        d_model=128,
        nhead=4,
        num_layers=2,
        dim_feedforward=256,
        gru_hidden=64,
        gru_layers=1,
        head_hidden=128,
        dropout=0.30,
        num_classes=4,
    ):
        super().__init__()

        if d_model % nhead != 0:
            raise ValueError(f"d_model ({d_model}) must be divisible by nhead ({nhead})")

        self.clinical_raw_dim = int(clinical_raw_dim)
        self.mri_feat_dim = int(mri_feat_dim)
        self.flag_dim = int(flag_dim)
        self.raw_total_dim = self.clinical_raw_dim + self.mri_feat_dim + self.flag_dim

        clinical_input_dim = self.clinical_raw_dim * 2       # scaled + mask
        mri_fusion_dim = (self.mri_feat_dim + self.flag_dim) * 2  # scaled + mask

        # ---- Branche Transformer (clinique) ----
        self.input_proj = nn.Sequential(
            nn.Linear(clinical_input_dim, d_model),
            nn.LayerNorm(d_model),
            nn.GELU(),
            nn.Dropout(dropout)
        )
        self.pos = nn.Parameter(torch.randn(1, max_len, d_model) * 0.02)

        enc_layer = nn.TransformerEncoderLayer(
            d_model=d_model,
            nhead=nhead,
            dim_feedforward=dim_feedforward,
            dropout=dropout,
            batch_first=True,
            activation="gelu",
            norm_first=False
        )
        self.encoder = nn.TransformerEncoder(enc_layer, num_layers=num_layers)
        self.tr_norm = nn.LayerNorm(d_model)

        self.tr_attn = nn.Sequential(
            nn.Linear(d_model, d_model // 2),
            nn.Tanh(),
            nn.Dropout(dropout),
            nn.Linear(d_model // 2, 1)
        )

        # ---- Branche BiGRU (clinique, EN PARALLELE du Transformer) ----
        self.gru = nn.GRU(
            input_size=clinical_input_dim,
            hidden_size=gru_hidden,
            num_layers=gru_layers,
            batch_first=True,
            dropout=(dropout if gru_layers > 1 else 0.0),
            bidirectional=True
        )
        self.gru_norm = nn.LayerNorm(gru_hidden * 2)

        # ---- Branche MRI/image (dernier point temporel uniquement) ----
        self.mri_branch = nn.Sequential(
            nn.Linear(mri_fusion_dim, 128),
            nn.LayerNorm(128),
            nn.GELU(),
            nn.Dropout(dropout),
            nn.Linear(128, 64),
            nn.GELU(),
        )

        # ---- Fusion finale (3 vecteurs) ----
        fusion_dim = d_model + (gru_hidden * 2) + 64
        self.head = nn.Sequential(
            nn.Linear(fusion_dim, head_hidden),
            nn.LayerNorm(head_hidden),
            nn.GELU(),
            nn.Dropout(dropout),
            nn.Linear(head_hidden, num_classes)
        )

    def split_clinical_mri(self, x):
        C = self.clinical_raw_dim
        M = self.mri_feat_dim
        Fl = self.flag_dim
        R = self.raw_total_dim

        scaled = x[:, :, :R]
        masks = x[:, :, R:R + R]

        clinical_scaled = scaled[:, :, :C]
        clinical_mask = masks[:, :, :C]
        x_clinical = torch.cat([clinical_scaled, clinical_mask], dim=-1)

        mri_start = C
        mri_end = C + M + Fl
        mri_scaled_last = scaled[:, -1, mri_start:mri_end]
        mri_mask_last = masks[:, -1, mri_start:mri_end]
        x_mri = torch.cat([mri_scaled_last, mri_mask_last], dim=-1)

        return x_clinical, x_mri

    def forward(self, x):
        x_clinical, x_mri = self.split_clinical_mri(x)

        # --- Branche Transformer ---
        h = self.input_proj(x_clinical)
        h = h + self.pos[:, :x.size(1), :]
        h = self.encoder(h)
        h = self.tr_norm(h)
        tr_scores = self.tr_attn(h).squeeze(-1)
        tr_w = torch.softmax(tr_scores, dim=1)
        tr_ctx = torch.sum(h * tr_w.unsqueeze(-1), dim=1)          # vecteur 1

        # --- Branche BiGRU (EN PARALLELE, reçoit x_clinical directement) ---
        g, _ = self.gru(x_clinical)
        g_last = self.gru_norm(g[:, -1, :])                        # vecteur 2

        # --- Branche MRI ---
        mri_ctx = self.mri_branch(x_mri)                           # vecteur 3

        # --- Fusion finale à 3 vecteurs ---
        z = torch.cat([tr_ctx, g_last, mri_ctx], dim=1)
        logits = self.head(z)
        return logits

# ==========================================================
# 15) HORIZON DEFAULTS (nb epochs / patience finaux par horizon)
# ==========================================================
HORIZON_DEFAULTS = {
    "T2": {"max_epochs": 60, "patience": 8,  "focal_gamma": 1.5},
    "T4": {"max_epochs": 60, "patience": 10, "focal_gamma": 2.0},
    "T6": {"max_epochs": 70, "patience": 10, "focal_gamma": 2.0},
}

SEARCH_EPOCHS = 20
SEARCH_PATIENCE = 5


def plot_losses(tr, va, title):
    plt.figure(figsize=(6, 4))
    plt.plot(tr, label="train_loss")
    plt.plot(va, label="val_loss")
    plt.title(title)
    plt.xlabel("epoch")
    plt.ylabel("loss")
    plt.legend()
    plt.tight_layout()
    plt.show()


# ==========================================================
# 16) TRAIN ONE MODEL (Transformer//BiGRU parallèle + MRI late-fusion)
# ==========================================================
def train_levels_model(
    Xtr, ytr, Xva, yva,
    clinical_raw_dim, mri_feat_dim, flag_dim, max_len,
    d_model, nhead, num_layers, dim_feedforward,
    gru_hidden, gru_layers, head_hidden, dropout,
    lr, weight_decay, batch_size,
    max_epochs, patience,
    aug_noise, input_drop_p, focal_gamma,
    grad_clip=1.0, num_classes=4
):
    Xtr = Xtr.astype(np.float32, copy=False)
    Xva = Xva.astype(np.float32, copy=False)

    Xtr_aug = augment_noise(Xtr, scale=aug_noise)
    Xtr_aug = apply_input_dropout(Xtr_aug, p=input_drop_p, fill_value=0.0)

    class_weights = inverse_freq_weights(ytr, num_classes)
    criterion = FocalLossMulti(gamma=focal_gamma, weight=class_weights, reduction="mean")

    model = LevelsParallelLateFusionModel(
        clinical_raw_dim=clinical_raw_dim,
        mri_feat_dim=mri_feat_dim,
        flag_dim=flag_dim,
        max_len=max_len,
        d_model=d_model,
        nhead=nhead,
        num_layers=num_layers,
        dim_feedforward=dim_feedforward,
        gru_hidden=gru_hidden,
        gru_layers=gru_layers,
        head_hidden=head_hidden,
        dropout=dropout,
        num_classes=num_classes,
    ).to(device)

    optimizer = torch.optim.AdamW(model.parameters(), lr=lr, weight_decay=weight_decay)
    scheduler = torch.optim.lr_scheduler.ReduceLROnPlateau(optimizer, mode="max", factor=0.5, patience=3, min_lr=1e-5)

    sampler = make_sampler(ytr, num_classes)

    train_ds = TensorDataset(torch.tensor(Xtr_aug, dtype=torch.float32), torch.tensor(ytr, dtype=torch.long))
    val_ds = TensorDataset(torch.tensor(Xva, dtype=torch.float32), torch.tensor(yva, dtype=torch.long))

    train_dl = DataLoader(train_ds, batch_size=batch_size, sampler=sampler, num_workers=NUM_WORKERS, pin_memory=PIN_MEMORY)
    val_dl = DataLoader(val_ds, batch_size=max(128, batch_size * 2), shuffle=False, num_workers=NUM_WORKERS, pin_memory=PIN_MEMORY)

    best_score = -np.inf
    best_state = None
    wait = 0
    tr_losses, va_losses = [], []

    for ep in range(1, max_epochs + 1):
        model.train()
        total = 0.0
        for xb, yb in train_dl:
            xb, yb = xb.to(device), yb.to(device)
            optimizer.zero_grad(set_to_none=True)
            logits = model(xb)
            loss = criterion(logits, yb)
            loss.backward()
            torch.nn.utils.clip_grad_norm_(model.parameters(), grad_clip)
            optimizer.step()
            total += loss.item()
        train_loss = total / max(len(train_dl), 1)

        model.eval()
        val_total = 0.0
        y_true_all, y_pred_all = [], []
        with torch.no_grad():
            for xb, yb in val_dl:
                xb, yb = xb.to(device), yb.to(device)
                logits = model(xb)
                loss = criterion(logits, yb)
                val_total += loss.item()
                pred = logits.argmax(dim=1)
                y_true_all.extend(yb.cpu().numpy().tolist())
                y_pred_all.extend(pred.cpu().numpy().tolist())

        val_loss = val_total / max(len(val_dl), 1)
        f1m = f1_score(y_true_all, y_pred_all, average="macro", zero_division=0)
        bacc = balanced_accuracy_score(y_true_all, y_pred_all)
        val_score = float(np.nanmean([f1m, bacc]))

        scheduler.step(val_score)
        tr_losses.append(train_loss)
        va_losses.append(val_loss)

        if ep == 1 or ep % 5 == 0:
            lr_now = optimizer.param_groups[0]["lr"]
            print(f"[Levels] ep{ep:02d} lr={lr_now:.2e} train_loss={train_loss:.5f} val_loss={val_loss:.5f} val_f1={f1m:.4f} val_bacc={bacc:.4f}")

        if val_score > best_score:
            best_score = val_score
            best_state = {k: v.detach().cpu().clone() for k, v in model.state_dict().items()}
            wait = 0
        else:
            wait += 1
            if wait >= patience:
                print("✅ Early stopping triggered.")
                break

    if best_state is not None:
        model.load_state_dict(best_state)

    model.eval()
    return model, tr_losses, va_losses, best_score


# ==========================================================
# 17) OPTUNA OBJECTIVE (pas de fold : split train/val du seed courant)
# ==========================================================
def optuna_objective(trial, Xtr, ytr, Xva, yva, clinical_raw_dim, mri_feat_dim, flag_dim, max_len, focal_gamma_default):
    d_model = trial.suggest_categorical("d_model", [64, 96, 128])
    possible_heads = [2, 4, 8]
    valid_heads = [h for h in possible_heads if d_model % h == 0]
    nhead = trial.suggest_categorical("nhead", valid_heads)
    num_layers = trial.suggest_int("num_layers", 1, 3)
    dim_feedforward = trial.suggest_categorical("dim_feedforward", [192, 256, 320])
    gru_hidden = trial.suggest_categorical("gru_hidden", [64, 96, 128])
    gru_layers = trial.suggest_int("gru_layers", 1, 2)
    head_hidden = trial.suggest_categorical("head_hidden", [128, 160, 192])
    dropout = trial.suggest_float("dropout", 0.10, 0.35)
    lr = trial.suggest_float("lr", 2e-4, 1e-3, log=True)
    weight_decay = trial.suggest_float("weight_decay", 1e-6, 5e-4, log=True)
    aug_noise = trial.suggest_float("aug_noise", 0.0, 0.02)
    input_drop_p = trial.suggest_float("input_drop_p", 0.0, 0.05)
    focal_gamma = trial.suggest_float("focal_gamma", max(0.5, focal_gamma_default - 0.7), focal_gamma_default + 0.7)

    _, _, _, best_score = train_levels_model(
        Xtr, ytr, Xva, yva,
        clinical_raw_dim=clinical_raw_dim, mri_feat_dim=mri_feat_dim, flag_dim=flag_dim, max_len=max_len,
        d_model=d_model, nhead=nhead, num_layers=num_layers, dim_feedforward=dim_feedforward,
        gru_hidden=gru_hidden, gru_layers=gru_layers, head_hidden=head_hidden, dropout=dropout,
        lr=lr, weight_decay=weight_decay, batch_size=32,
        max_epochs=SEARCH_EPOCHS, patience=SEARCH_PATIENCE,
        aug_noise=aug_noise, input_drop_p=input_drop_p, focal_gamma=focal_gamma,
        grad_clip=GRAD_CLIP, num_classes=NUM_CLASSES
    )

    trial.report(float(best_score), step=0)
    if trial.should_prune():
        raise optuna.exceptions.TrialPruned()

    return float(best_score)


# ==========================================================
# 18) RUN ONE HORIZON (UN SEED, SANS FOLD)
# ==========================================================
def run_one_horizon_levels(X, y, pat, name="T2", n_trials=20, seed=42):
    print("\n" + "=" * 70)
    print(f"RUN PARALLEL TRANSFORMER // BiGRU + MRI LATE-FUSION (3 branches) - {name} - seed {seed}")
    print("=" * 70)

    set_seed(seed)
    defaults = HORIZON_DEFAULTS[name]

    (X_tr, y_tr, pat_tr), (X_va, y_va, pat_va), (X_te, y_te, pat_te) = split_train_val_test_patientwise(
        X, y, pat, TRAIN_RATIO, VAL_RATIO, TEST_RATIO, seed=seed
    )
    print_split_info(name, (X_tr, y_tr, pat_tr), (X_va, y_va, pat_va), (X_te, y_te, pat_te))

    X_tr_pp, med, scaler = fit_preprocess_train_only_with_mask(X_tr)
    X_va_pp = transform_with_train_params_with_mask(X_va, med, scaler)
    X_te_pp = transform_with_train_params_with_mask(X_te, med, scaler)

    max_len = X_tr_pp.shape[1]

    sampler_opt = TPESampler(seed=seed)
    pruner = MedianPruner(n_startup_trials=5)
    study = optuna.create_study(direction="maximize", sampler=sampler_opt, pruner=pruner)

    study.optimize(
        lambda trial: optuna_objective(
            trial, X_tr_pp, y_tr, X_va_pp, y_va,
            CLINICAL_RAW_DIM, MRI_FEAT_DIM, FLAG_DIM, max_len,
            defaults["focal_gamma"]
        ),
        n_trials=n_trials,
        show_progress_bar=True
    )

    best = study.best_params
    print("\n✅ BEST PARAMS:", best)

    model, tr_loss, va_loss, best_val_score = train_levels_model(
        X_tr_pp, y_tr, X_va_pp, y_va,
        clinical_raw_dim=CLINICAL_RAW_DIM, mri_feat_dim=MRI_FEAT_DIM, flag_dim=FLAG_DIM, max_len=max_len,
        d_model=best["d_model"], nhead=best["nhead"], num_layers=best["num_layers"],
        dim_feedforward=best["dim_feedforward"], gru_hidden=best["gru_hidden"], gru_layers=best["gru_layers"],
        head_hidden=best["head_hidden"], dropout=best["dropout"],
        lr=best["lr"], weight_decay=best["weight_decay"], batch_size=64,
        max_epochs=defaults["max_epochs"], patience=defaults["patience"],
        aug_noise=best["aug_noise"], input_drop_p=best["input_drop_p"], focal_gamma=best["focal_gamma"],
        grad_clip=GRAD_CLIP, num_classes=NUM_CLASSES
    )

    plot_losses(tr_loss, va_loss, f"{name} (seed {seed}) - Parallel Transformer//BiGRU + MRI late-fusion")

    # ---- TEST ----
    model.eval()
    test_ds = TensorDataset(torch.tensor(X_te_pp, dtype=torch.float32))
    test_dl = DataLoader(test_ds, batch_size=128, shuffle=False, num_workers=NUM_WORKERS, pin_memory=PIN_MEMORY)

    preds = []
    with torch.no_grad():
        for (xb,) in test_dl:
            xb = xb.to(device)
            logits = model(xb)
            preds.extend(logits.argmax(dim=1).cpu().numpy().tolist())
    preds = np.array(preds, dtype=np.int64)

    test_acc = accuracy_score(y_te, preds)
    test_precision = precision_score(y_te, preds, average="macro", zero_division=0)
    test_recall = recall_score(y_te, preds, average="macro", zero_division=0)
    test_f1 = f1_score(y_te, preds, average="macro", zero_division=0)
    test_bacc = balanced_accuracy_score(y_te, preds)
    cm = confusion_matrix(y_te, preds, labels=[0, 1, 2, 3])
    per_class_report = classification_report(
        y_te, preds, labels=[0, 1, 2, 3], target_names=CLASS_NAMES,
        output_dict=True, zero_division=0
    )

    print("\n=== TEST REPORT ===")
    print(f"Accuracy: {test_acc:.4f} | Precision: {test_precision:.4f} | Recall: {test_recall:.4f} | F1: {test_f1:.4f} | BAcc: {test_bacc:.4f}")
    print("Confusion Matrix:\n", cm)
    print(classification_report(y_te, preds, labels=[0, 1, 2, 3], target_names=CLASS_NAMES, zero_division=0))

    return {
        "name": name,
        "seed": seed,
        "accuracy": float(test_acc),
        "precision": float(test_precision),
        "recall": float(test_recall),
        "f1": float(test_f1),
        "balanced_accuracy": float(test_bacc),
        "confusion_matrix": cm.tolist(),
        "per_class": per_class_report,
        "best_params": best,
        "n_pat_train": int(len(np.unique(pat_tr))),
        "n_pat_val": int(len(np.unique(pat_va))),
        "n_pat_test": int(len(np.unique(pat_te))),
        "n_seq_train": int(X_tr.shape[0]),
        "n_seq_val": int(X_va.shape[0]),
        "n_seq_test": int(X_te.shape[0]),
        "train_loss_curve": [float(v) for v in tr_loss],
        "val_loss_curve": [float(v) for v in va_loss],
        "final_train_loss": float(tr_loss[-1]) if len(tr_loss) else float("nan"),
        "final_val_loss": float(va_loss[-1]) if len(va_loss) else float("nan"),
    }

# ==========================================================
# 19) RUN ALL WITH MULTI-SEED (SANS FOLD)
# ==========================================================
def mean_std(x):
    if len(x) == 0:
        return 0.0, 0.0
    return float(np.mean(x)), float(np.std(x))


all_results = {}

for horizon_name, (X, y, p) in {"T2": (X2, y2, p2), "T4": (X4, y4, p4), "T6": (X6, y6, p6)}.items():

    print("\n" + "=" * 80)
    print(f"RUNNING {horizon_name}")
    print("=" * 80)

    acc_scores, precision_scores, recall_scores, f1_scores, bacc_scores = [], [], [], [], []
    confusion_matrices = []
    per_class_seed_results = []
    all_seed_results = []
    train_loss_curves_seeds, val_loss_curves_seeds = [], []
    final_train_losses, final_val_losses = [], []

    for seed in SEEDS:
        print("\n" + "-" * 60)
        print(f"SEED = {seed}")
        print("-" * 60)

        res = run_one_horizon_levels(X, y, p, name=horizon_name, n_trials=N_TRIALS, seed=seed)

        print(f"\nTEST RESULTS FOR SEED = {seed}:")
        print(f"Accuracy = {res['accuracy']:.4f} | Precision = {res['precision']:.4f} | "
              f"Recall = {res['recall']:.4f} | F1 = {res['f1']:.4f} | BAcc = {res['balanced_accuracy']:.4f}")
        print("Confusion Matrix:\n", np.array(res["confusion_matrix"]))

        acc_scores.append(res["accuracy"])
        precision_scores.append(res["precision"])
        recall_scores.append(res["recall"])
        f1_scores.append(res["f1"])
        bacc_scores.append(res["balanced_accuracy"])
        confusion_matrices.append(np.array(res["confusion_matrix"]))
        per_class_seed_results.append(res["per_class"])
        all_seed_results.append(res)

        train_loss_curves_seeds.append(res["train_loss_curve"])
        val_loss_curves_seeds.append(res["val_loss_curve"])
        final_train_losses.append(res["final_train_loss"])
        final_val_losses.append(res["final_val_loss"])

    # ---- Moyennes ± std globales ----
    mean_acc, std_acc = mean_std(acc_scores)
    mean_prec, std_prec = mean_std(precision_scores)
    mean_rec, std_rec = mean_std(recall_scores)
    mean_f1, std_f1 = mean_std(f1_scores)
    mean_bacc, std_bacc = mean_std(bacc_scores)

    print("\n" + "=" * 80)
    print(f"AVERAGE RESULTS - {horizon_name}")
    print("=" * 80)
    print(f"Accuracy: {mean_acc:.4f} ± {std_acc:.4f}")
    print(f"Precision: {mean_prec:.4f} ± {std_prec:.4f}")
    print(f"Recall: {mean_rec:.4f} ± {std_rec:.4f}")
    print(f"F1-score: {mean_f1:.4f} ± {std_f1:.4f}")
    print(f"Balanced Accuracy: {mean_bacc:.4f} ± {std_bacc:.4f}")

    # ---- Per-class mean ± std ----
    per_class_summary = {}
    print("\nPer-class results (mean ± std across seeds):")
    for cls_name in CLASS_NAMES:
        cls_prec = [r[cls_name]["precision"] for r in per_class_seed_results if cls_name in r]
        cls_rec = [r[cls_name]["recall"] for r in per_class_seed_results if cls_name in r]
        cls_f1 = [r[cls_name]["f1-score"] for r in per_class_seed_results if cls_name in r]
        cls_sup = [r[cls_name]["support"] for r in per_class_seed_results if cls_name in r]

        mp, sp = mean_std(cls_prec)
        mr, sr = mean_std(cls_rec)
        mf, sf = mean_std(cls_f1)
        ms, ss = mean_std(cls_sup)

        per_class_summary[cls_name] = {
            "precision": f"{mp:.4f} ± {sp:.4f}",
            "recall": f"{mr:.4f} ± {sr:.4f}",
            "f1-score": f"{mf:.4f} ± {sf:.4f}",
            "support": f"{ms:.1f} ± {ss:.1f}",
        }
        print(f"\n{cls_name}")
        print(f"  Precision: {mp:.4f} ± {sp:.4f}")
        print(f"  Recall   : {mr:.4f} ± {sr:.4f}")
        print(f"  F1-score : {mf:.4f} ± {sf:.4f}")
        print(f"  Support  : {ms:.1f} ± {ss:.1f}")

    # ---- Matrice de confusion moyenne (arrondie) ----
    avg_cm = np.mean(np.stack(confusion_matrices), axis=0)
    std_cm = np.std(np.stack(confusion_matrices), axis=0)
    avg_cm_rounded = np.round(avg_cm, 1)

    print("\nAverage Confusion Matrix (mean across seeds, arrondie) :")
    print(avg_cm_rounded)
    print("\nStd Confusion Matrix:")
    print(np.round(std_cm, 1))

    # ---- Courbes de loss moyennes ----
    def pad_curve_to_length(curve, length):
        curve = list(curve)
        if len(curve) == 0:
            return [np.nan] * length
        if len(curve) >= length:
            return curve[:length]
        return curve + [curve[-1]] * (length - len(curve))

    def mean_std_curves(curves):
        curves = [c for c in curves if len(c) > 0]
        if len(curves) == 0:
            return np.array([]), np.array([])
        max_len_c = max(len(c) for c in curves)
        padded = np.array([pad_curve_to_length(c, max_len_c) for c in curves], dtype=float)
        return np.nanmean(padded, axis=0), np.nanstd(padded, axis=0)

    mean_train_loss_curve, std_train_loss_curve = mean_std_curves(train_loss_curves_seeds)
    mean_val_loss_curve, std_val_loss_curve = mean_std_curves(val_loss_curves_seeds)
    mean_final_train_loss, std_final_train_loss = mean_std(final_train_losses)
    mean_final_val_loss, std_final_val_loss = mean_std(final_val_losses)

    print("\nMean Train Loss / Mean Validation Loss (5 seeds):")
    print(f"  Mean Train Loss (dernière epoch) : {mean_final_train_loss:.5f} ± {std_final_train_loss:.5f}")
    print(f"  Mean Val Loss   (dernière epoch) : {mean_final_val_loss:.5f} ± {std_final_val_loss:.5f}")

    all_results[horizon_name] = {
        "Accuracy": f"{mean_acc:.4f} ± {std_acc:.4f}",
        "Precision": f"{mean_prec:.4f} ± {std_prec:.4f}",
        "Recall": f"{mean_rec:.4f} ± {std_rec:.4f}",
        "F1-score": f"{mean_f1:.4f} ± {std_f1:.4f}",
        "Balanced Accuracy": f"{mean_bacc:.4f} ± {std_bacc:.4f}",
        "Per-class": per_class_summary,
        "Mean_confusion_matrix": avg_cm_rounded.tolist(),
        "Std_confusion_matrix": np.round(std_cm, 1).tolist(),
        "Mean_Train_Loss_Final": f"{mean_final_train_loss:.5f} ± {std_final_train_loss:.5f}",
        "Mean_Val_Loss_Final": f"{mean_final_val_loss:.5f} ± {std_final_val_loss:.5f}",
        "Mean_Train_Loss_Curve": mean_train_loss_curve.tolist(),
        "Std_Train_Loss_Curve": std_train_loss_curve.tolist(),
        "Mean_Val_Loss_Curve": mean_val_loss_curve.tolist(),
        "Std_Val_Loss_Curve": std_val_loss_curve.tolist(),
        "Per_seed_results": all_seed_results,
    }


# ==========================================================
# FINAL RESULTS (résumé console)
# ==========================================================
print("\n" + "=" * 80)
print("FINAL RESULTS")
print("=" * 80)
for horizon, metrics in all_results.items():
    print("\n" + "-" * 50)
    print(horizon)
    print("-" * 50)
    for metric, value in metrics.items():
        if metric not in ("Per_seed_results", "Mean_Train_Loss_Curve", "Std_Train_Loss_Curve",
                           "Mean_Val_Loss_Curve", "Std_Val_Loss_Curve",
                           "Mean_confusion_matrix", "Std_confusion_matrix"):
            print(f"{metric}: {value}")


# ==========================================================
# 20) FIGURE COMBINEE - 3 MATRICES DE CONFUSION (T2 / T4 / T6)
# ==========================================================
HORIZON_ORDER = [h for h in ["T2", "T4", "T6"] if h in all_results]

fig_cm, axes_cm = plt.subplots(1, len(HORIZON_ORDER), figsize=(7 * len(HORIZON_ORDER), 6))
if len(HORIZON_ORDER) == 1:
    axes_cm = [axes_cm]

short_names = ["Minimal", "Mild", "Moderate", "Severe"]

for ax, horizon_name in zip(axes_cm, HORIZON_ORDER):
    avg_cm = np.array(all_results[horizon_name]["Mean_confusion_matrix"])
    im = ax.imshow(avg_cm, cmap="Blues")
    ax.set_title(f"{horizon_name} - Mean Confusion Matrix (5 seeds)")
    ax.set_xlabel("Predicted")
    ax.set_ylabel("True")
    ax.set_xticks(range(len(short_names)))
    ax.set_xticklabels(short_names, rotation=45, ha="right")
    ax.set_yticks(range(len(short_names)))
    ax.set_yticklabels(short_names)
    for i in range(avg_cm.shape[0]):
        for j in range(avg_cm.shape[1]):
            ax.text(j, i, f"{avg_cm[i, j]:.1f}", ha="center", va="center",
                     color="white" if avg_cm[i, j] > avg_cm.max() / 2 else "black")
    fig_cm.colorbar(im, ax=ax, fraction=0.046, pad=0.04)

fig_cm.suptitle("Matrices de confusion moyennes arrondies (mean across 5 seeds) - T2 / T4 / T6")
fig_cm.tight_layout()

cm_fig_path = OUT_DIR / "confusion_matrices_T2_T4_T6.png"
fig_cm.savefig(cm_fig_path, dpi=200, bbox_inches="tight")
plt.show()
print("✅ Figure des matrices de confusion sauvegardée :", cm_fig_path)


# ==========================================================
# 21) FIGURE COMBINEE - 3 COURBES TRAIN LOSS / VAL LOSS (T2 / T4 / T6)
# ==========================================================
fig_loss, axes_loss = plt.subplots(1, len(HORIZON_ORDER), figsize=(6 * len(HORIZON_ORDER), 4.5))
if len(HORIZON_ORDER) == 1:
    axes_loss = [axes_loss]

for ax, horizon_name in zip(axes_loss, HORIZON_ORDER):
    mtr = np.array(all_results[horizon_name]["Mean_Train_Loss_Curve"])
    mva = np.array(all_results[horizon_name]["Mean_Val_Loss_Curve"])
    str_ = np.array(all_results[horizon_name]["Std_Train_Loss_Curve"])
    sva = np.array(all_results[horizon_name]["Std_Val_Loss_Curve"])

    if len(mtr) > 0:
        epochs = np.arange(1, len(mtr) + 1)
        ax.plot(epochs, mtr, label="Mean Train Loss", color="tab:blue")
        ax.fill_between(epochs, mtr - str_, mtr + str_, color="tab:blue", alpha=0.15)
        ax.plot(epochs, mva, label="Mean Val Loss", color="tab:orange")
        ax.fill_between(epochs, mva - sva, mva + sva, color="tab:orange", alpha=0.15)

    ax.set_title(f"{horizon_name} - Train/Val Loss (mean ± std, 5 seeds)")
    ax.set_xlabel("epoch")
    ax.set_ylabel("loss")
    ax.legend()

fig_loss.suptitle("Courbes Train Loss / Validation Loss moyennes (mean across 5 seeds) - T2 / T4 / T6")
fig_loss.tight_layout()

loss_fig_path = OUT_DIR / "train_val_loss_curves_T2_T4_T6.png"
fig_loss.savefig(loss_fig_path, dpi=200, bbox_inches="tight")
plt.show()
print("✅ Figure des courbes train/val loss sauvegardée :", loss_fig_path)


# ==========================================================
# 22) TABLEAU FINAL COMPLET DE TOUTES LES METRIQUES
# ==========================================================
final_rows = []
for horizon_name in HORIZON_ORDER:
    m = all_results[horizon_name]
    row = {
        "Horizon": horizon_name,
        "Mean Accuracy ± Std": m["Accuracy"],
        "Mean Precision ± Std": m["Precision"],
        "Mean Recall ± Std": m["Recall"],
        "Mean F1 ± Std": m["F1-score"],
        "Mean Balanced Accuracy ± Std": m["Balanced Accuracy"],
        "Mean Train Loss ± Std": m["Mean_Train_Loss_Final"],
        "Mean Validation Loss ± Std": m["Mean_Val_Loss_Final"],
    }
    for cls_name in CLASS_NAMES:
        cls_metrics = m["Per-class"][cls_name]
        row[f"{cls_name} - Precision (mean ± std)"] = cls_metrics["precision"]
        row[f"{cls_name} - Recall (mean ± std)"] = cls_metrics["recall"]
        row[f"{cls_name} - F1-score (mean ± std)"] = cls_metrics["f1-score"]
        row[f"{cls_name} - Support (mean ± std)"] = cls_metrics["support"]
    final_rows.append(row)

final_metrics_df = pd.DataFrame(final_rows)

print("\n" + "=" * 80)
print("TABLEAU FINAL COMPLET DE TOUTES LES METRIQUES")
print("=" * 80)
display(final_metrics_df)


# ==========================================================
# 23) EXPORTS AUTOMATIQUES : JSON / CSV / EXCEL
# ==========================================================
results_json_path = OUT_DIR / "final_results_multi_seed_parallel_late_fusion.json"
with open(results_json_path, "w", encoding="utf-8") as f:
    json.dump(all_results, f, indent=2, ensure_ascii=False)
print("✅ Export JSON sauvegardé :", results_json_path)

rows = []
for horizon, metrics in all_results.items():
    rows.append({
        "Horizon": horizon,
        "Accuracy": metrics["Accuracy"],
        "Precision": metrics["Precision"],
        "Recall": metrics["Recall"],
        "F1-score": metrics["F1-score"],
        "Balanced Accuracy": metrics["Balanced Accuracy"],
        "Mean Train Loss": metrics["Mean_Train_Loss_Final"],
        "Mean Validation Loss": metrics["Mean_Val_Loss_Final"],
    })
summary_df = pd.DataFrame(rows)
summary_csv_path = OUT_DIR / "summary_results_multi_seed_parallel_late_fusion.csv"
summary_df.to_csv(summary_csv_path, index=False)
print("✅ Export CSV (résumé) sauvegardé :", summary_csv_path)

final_csv_path = OUT_DIR / "final_metrics_table_complete.csv"
final_metrics_df.to_csv(final_csv_path, index=False)
print("✅ Export CSV (complet, par classe) sauvegardé :", final_csv_path)

final_xlsx_path = OUT_DIR / "final_metrics_table_complete.xlsx"
try:
    with pd.ExcelWriter(final_xlsx_path, engine="openpyxl") as writer:
        final_metrics_df.to_excel(writer, sheet_name="Resume_complet", index=False)
        summary_df.to_excel(writer, sheet_name="Resume_court", index=False)

        keep_cols = [
            "seed", "accuracy", "precision", "recall", "f1", "balanced_accuracy",
            "n_pat_train", "n_pat_val", "n_pat_test",
            "n_seq_train", "n_seq_val", "n_seq_test",
            "final_train_loss", "final_val_loss"
        ]
        for horizon_name in HORIZON_ORDER:
            per_seed_df = pd.DataFrame(all_results[horizon_name]["Per_seed_results"])
            cols_present = [c for c in keep_cols if c in per_seed_df.columns]
            per_seed_df[cols_present].to_excel(writer, sheet_name=f"{horizon_name}_par_seed", index=False)
    print("✅ Export Excel sauvegardé :", final_xlsx_path)
except Exception as e:
    print("⚠️ Export Excel impossible (openpyxl manquant ?) :", e)


# ==========================================================
# 24) AFFICHAGE AUTOMATIQUE FINAL
# ==========================================================
print("\n" + "=" * 80)
print("RECAPITULATIF DES FICHIERS EXPORTES")
print("=" * 80)
print("✅ JSON   :", results_json_path)
print("✅ CSV    :", summary_csv_path)
print("✅ CSV    :", final_csv_path)
print("✅ Excel  :", final_xlsx_path)
print("✅ PNG    :", cm_fig_path)
print("✅ PNG    :", loss_fig_path)

print("\n" + "=" * 80)
print("RESUME COURT (Accuracy / Precision / Recall / F1 / Balanced Accuracy)")
print("=" * 80)
display(summary_df)

print("\n" + "=" * 80)
print("TABLEAU COMPLET (avec métriques par classe + loss)")
print("=" * 80)
display(final_metrics_df)
```
