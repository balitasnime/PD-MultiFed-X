```python
import re
import json
import random
import numpy as np
import pandas as pd
from pathlib import Path
from functools import reduce

import matplotlib.pyplot as plt

from sklearn.model_selection import GroupShuffleSplit
from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import (
    classification_report,
    confusion_matrix,
    f1_score,
    balanced_accuracy_score,
    accuracy_score,
    precision_score,
    recall_score
)

import torch
import torch.nn as nn
from torch.utils.data import DataLoader, TensorDataset

import optuna
from optuna.samplers import TPESampler
from optuna.pruners import MedianPruner


# ==========================================================
# 0) CONFIG
# ==========================================================
OUT_DIR = Path("/kaggle/working")
OUT_DIR.mkdir(parents=True, exist_ok=True)

# ✅ FORCER CPU pour éviter l'erreur CUDA/P100
device = torch.device("cpu")
print("device:", device)

TRAIN_RATIO = 0.70
VAL_RATIO   = 0.15
TEST_RATIO  = 0.15
assert abs(TRAIN_RATIO + VAL_RATIO + TEST_RATIO - 1.0) < 1e-9

N_TRIALS = 20

# ✅ MRI/image features for late fusion
MRI_FEAT_DIM = 256
MRI_MISSING_FILL = -1.0

# ✅ CPU
USE_AMP = False
USE_COMPILE = False
NUM_WORKERS = 2
PIN_MEMORY = False


# ==========================================================
# 1) SEED
# ==========================================================
def set_seed(seed=42):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    if torch.cuda.is_available():
        torch.cuda.manual_seed_all(seed)
    if hasattr(torch.backends, "cudnn"):
        torch.backends.cudnn.deterministic = True
        torch.backends.cudnn.benchmark = False


# ==========================================================
# 2) FILE HELPERS
# ==========================================================
def find_file(filename):
    base = Path("/kaggle/input")
    matches = list(base.rglob(filename))
    if not matches:
        raise FileNotFoundError(f"{filename} introuvable dans /kaggle/input")
    return matches[0]

def find_mri_feature_file():
    """Find pre-extracted MRI/image featuress CSV in Kaggle input."""
    candidates = list(Path("/kaggle/input").rglob("mri_featuress.csv"))
    if len(candidates) == 0:
        raise FileNotFoundError("mri_featuress.csv introuvable dans /kaggle/input")
    print("[FOUND] MRI feature file ->", candidates[0])
    return candidates[0]

def read_csv_robust(path):
    return pd.read_csv(
        path,
        encoding="latin1",
        sep=None,
        engine="python",
        on_bad_lines="skip"
    )


# ==========================================================
# 3) MERGE HELPERS
# ==========================================================
KEYS = ["PATNO", "EVENT_ID"]
DATE_CANDIDATES = ["INFODT", "VISITDT", "DATE", "EXAMDATE", "TESTDATE"]

LABEL_COLS = ["GDS__DEPRESSION", "GDS__LEVELS", "UP23__STADE"]

ADMIN_WORDS = [
    "REC_ID", "PAG_NAME", "ORIG_ENTRY", "LAST_UPDATE",
    "ENTRY", "UPDATE", "NUPSOURC", "PTCGBOTH",
    "INFODT", "VISITDT", "EXAMDATE", "TESTDATE", "DATE"
]

LEAKY_CONTAINS = [
    "ESS_TOTAL", "SCORE_GDS", "SCAUTOT", "RBDTOT",
    "NP1RTOT", "NP2PTOT", "NP3TOT", "NHY"
]

_FR_MONTH = {
    "janv": "Jan", "févr": "Feb", "fevr": "Feb", "mars": "Mar", "avr": "Apr",
    "mai": "May", "juin": "Jun", "juil": "Jul", "août": "Aug", "aout": "Aug",
    "sept": "Sep", "oct": "Oct", "nov": "Nov", "déc": "Dec", "dec": "Dec"
}

def norm_cols(df):
    df = df.copy()
    df.columns = [str(c).strip().upper() for c in df.columns]
    return df

def coerce_keys(df):
    df = df.copy()
    if "PATNO" in df.columns:
        df["PATNO"] = pd.to_numeric(df["PATNO"], errors="coerce").astype("Int64")
    if "EVENT_ID" in df.columns:
        df["EVENT_ID"] = df["EVENT_ID"].astype(str).str.strip().str.upper()
    return df

def detect_date_col(df):
    for c in DATE_CANDIDATES:
        if c in df.columns:
            return c
    return None

def parse_date_series(s: pd.Series) -> pd.Series:
    x = s.astype(str).str.strip()
    for fr, en in _FR_MONTH.items():
        x = x.str.replace(fr, en, case=False, regex=False)
    return pd.to_datetime(x, errors="coerce", format="mixed", dayfirst=True)

def dedup_by_key_keep_latest(df):
    df = df.copy()
    date_col = detect_date_col(df)
    if date_col:
        df[date_col] = parse_date_series(df[date_col])
        df = df.sort_values(KEYS + [date_col]).drop_duplicates(subset=KEYS, keep="last")
    else:
        df = df.sort_values(KEYS).drop_duplicates(subset=KEYS, keep="first")
    return df

def prefix_features(df, prefix):
    df = norm_cols(df)
    df = coerce_keys(df)
    df = dedup_by_key_keep_latest(df)

    date_col = detect_date_col(df)
    rename = {}
    for c in df.columns:
        if c in KEYS:
            continue
        if date_col and c == date_col:
            rename[c] = f"{prefix}__VISIT_DATE"
        else:
            rename[c] = f"{prefix}__{c}"
    return df.rename(columns=rename)

def event_id_to_month(e):
    if pd.isna(e):
        return np.nan
    s = str(e).strip().upper()
    if s == "BL":
        return 0
    m = re.match(r"V(\d+)", s)
    if m:
        v = int(m.group(1))
        return (v - 2) * 6
    return np.nan

def drop_admin_cols_safe(df):
    protected = set(["PATNO", "EVENT_ID", "MONTH", "VISIT_DATE"]).union(LABEL_COLS)
    cols = []
    for c in df.columns:
        if c in protected:
            continue
        if any(w in c.upper() for w in ADMIN_WORDS):
            cols.append(c)
    return df.drop(columns=cols, errors="ignore")

def drop_leaky_cols_safe(df):
    protected = set(["PATNO", "EVENT_ID", "MONTH", "VISIT_DATE"]).union(LABEL_COLS)
    cols = []
    for c in df.columns:
        if c in protected:
            continue
        if any(w in c.upper() for w in LEAKY_CONTAINS):
            cols.append(c)
    return df.drop(columns=cols, errors="ignore")

def filter_to_pd(df, pd_patnos):
    df = norm_cols(df)
    df = coerce_keys(df)
    df = df[df["PATNO"].notna()].copy()
    df["PATNO"] = df["PATNO"].astype(int)
    return df[df["PATNO"].isin(pd_patnos)].copy()


# ==========================================================
# 4) LOAD TABLES
# ==========================================================
demo  = read_csv_robust(find_file("Demographics.csv"))
ess   = read_csv_robust(find_file("Epworth_Sleepiness_Scale.csv"))
up1   = read_csv_robust(find_file("MDS-UPDRS_Part_I.csv"))
up4   = read_csv_robust(find_file("MDS-UPDRS_Part_IV.csv"))
rem   = read_csv_robust(find_file("REM_Sleep_Behavior.csv"))
scopa = read_csv_robust(find_file("SCOPA-AUT.csv"))
gds   = read_csv_robust(find_file("Geriatric_Depression_Scale.csv"))
up23  = pd.read_excel(find_file("UPDRS_PartII_PartIII.xlsx"))
pstat = read_csv_robust(find_file("Participant_Status.csv"))

print(
    "✅ Loaded shapes:",
    demo.shape, ess.shape, up1.shape, up4.shape,
    rem.shape, scopa.shape, gds.shape, up23.shape, pstat.shape
)


# ==========================================================
# 5) PD PATNOS
# ==========================================================
pstat = norm_cols(pstat)
pstat = coerce_keys(pstat)

cohort_norm = (
    pstat["COHORT_DEFINITION"].astype(str).str.upper()
    .str.replace("’", "'", regex=False).str.strip()
)

pd_patnos = set(
    pstat.loc[cohort_norm.str.contains("PARKINSON"), "PATNO"]
    .dropna().astype(int).unique()
)
print("✅ PD PATNO count:", len(pd_patnos))


# ==========================================================
# 6) FILTER TABLES TO PD
# ==========================================================
demo  = filter_to_pd(demo, pd_patnos)
ess   = filter_to_pd(ess, pd_patnos)
up1   = filter_to_pd(up1, pd_patnos)
up4   = filter_to_pd(up4, pd_patnos)
rem   = filter_to_pd(rem, pd_patnos)
scopa = filter_to_pd(scopa, pd_patnos)
gds   = filter_to_pd(gds, pd_patnos)
up23  = filter_to_pd(up23, pd_patnos)

print(
    "✅ After PD filter shapes:",
    demo.shape, ess.shape, up1.shape, up4.shape,
    rem.shape, scopa.shape, gds.shape, up23.shape
)


# ==========================================================
# 7) MERGE (PATNO + EVENT_ID)
# ==========================================================
tables = {
    "DEMO": demo,
    "ESS": ess,
    "UP1": up1,
    "UP4": up4,
    "REM": rem,
    "SCOPA": scopa,
    "GDS": gds,
    "UP23": up23
}

prefixed = [prefix_features(df, name) for name, df in tables.items()]
merged = reduce(lambda a, b: pd.merge(a, b, on=KEYS, how="outer"), prefixed)
print("✅ merged shape:", merged.shape)

date_cols = [c for c in merged.columns if c.endswith("__VISIT_DATE")]
merged["VISIT_DATE"] = parse_date_series(merged[date_cols[0]]) if date_cols else pd.NaT

merged["MONTH"] = merged["EVENT_ID"].apply(event_id_to_month).astype("Int64")
merged = merged[merged["PATNO"].notna() & merged["MONTH"].notna()].copy()
merged = merged[merged["MONTH"] >= 0].copy()
merged = merged.sort_values(["PATNO", "MONTH", "VISIT_DATE"])
print("✅ months sample:", sorted(merged["MONTH"].unique())[:20])


# ==========================================================
# 8) CLEAN (NO IMPUTE/SCALE HERE to avoid leakage)
# ==========================================================
merged = drop_admin_cols_safe(merged)
merged = drop_leaky_cols_safe(merged)

protected = set(["PATNO", "EVENT_ID", "MONTH", "VISIT_DATE"])
protected_all = protected.union(LABEL_COLS)

for c in merged.columns:
    if c in protected_all:
        continue
    if merged[c].dtype == "O":
        conv = pd.to_numeric(merged[c], errors="coerce")
        if conv.notna().mean() >= 0.20:
            merged[c] = conv

DROP_MISS = 0.90
KEEP_PREFIX = ("ESS__", "REM__", "SCOPA__", "GDS__")
missing_ratio = merged.isna().mean()

drop_cols = []
for c in merged.columns:
    if c in protected_all:
        continue
    if c.startswith(KEEP_PREFIX):
        continue
    if missing_ratio[c] > DROP_MISS:
        drop_cols.append(c)

merged2 = merged.drop(columns=drop_cols).copy()
print("✅ after drop missing columns:", merged2.shape)

feature_cols_all = [c for c in merged2.columns if c not in protected_all]
num_cols = [c for c in feature_cols_all if pd.api.types.is_numeric_dtype(merged2[c])]
cat_cols = [c for c in feature_cols_all if c not in num_cols]

if cat_cols:
    merged2 = merged2.drop(columns=cat_cols, errors="ignore")

for c in LABEL_COLS:
    if c in merged2.columns:
        merged2[c] = pd.to_numeric(merged2[c], errors="coerce").round().astype("Int64")

df_seq = (
    merged2
    .sort_values(["PATNO", "MONTH", "VISIT_DATE"])
    .drop_duplicates(["PATNO", "MONTH"], keep="last")
)
print("✅ df_seq:", df_seq.shape)

dep_cols = sorted(set(
    [c for c in df_seq.columns if c.startswith("UP1__NP1")] +
    [c for c in df_seq.columns if c.startswith("ESS__")] +
    [c for c in df_seq.columns if c.startswith("GDS__") and c not in ["GDS__DEPRESSION", "GDS__LEVELS"]] +
    [c for c in df_seq.columns if c.startswith("REM__")] +
    [c for c in df_seq.columns if c.startswith("SCOPA__")]
))
dep_cols = [c for c in dep_cols if c not in protected_all and pd.api.types.is_numeric_dtype(df_seq[c])]
dep_cols = [c for c in dep_cols if not any(w in c.upper() for w in LEAKY_CONTAINS)]

print("✅ dep_cols:", len(dep_cols))

# ==========================================================
# 8B) LOAD MRI/IMAGE FEATURES + LATE FUSION INPUT
# ==========================================================
CLINICAL_RAW_DIM = len(dep_cols)

def load_preextracted_mri_features(mri_file):
    """
    Load mri_features.csv containing:
      - PATNO
      - mri_feat_0 ... mri_feat_255
      - optional mri_available
    """
    mri_df = pd.read_csv(mri_file)
    mri_df.columns = [str(c).strip() for c in mri_df.columns]

    if "PATNO" not in mri_df.columns:
        raise ValueError("MRI file must contain PATNO column")

    mri_df["PATNO"] = pd.to_numeric(mri_df["PATNO"], errors="coerce").astype("Int64")

    feat_cols = [c for c in mri_df.columns if c.lower().startswith("mri_feat")]
    feat_cols = sorted(feat_cols, key=lambda x: int(re.findall(r"\d+", x)[-1]) if re.findall(r"\d+", x) else 10**9)

    if len(feat_cols) != MRI_FEAT_DIM:
        raise ValueError(f"Expected {MRI_FEAT_DIM} MRI features, but found {len(feat_cols)}.")

    if "mri_available" not in mri_df.columns:
        mri_df["mri_available"] = 1

    mri_df = mri_df.dropna(subset=["PATNO"]).copy()
    mri_df["PATNO"] = mri_df["PATNO"].astype(int)

    feat_dict = {}
    for _, row in mri_df.iterrows():
        patno = int(row["PATNO"])
        available = float(row["mri_available"])

        if available > 0.5:
            feat = row[feat_cols].to_numpy(dtype=np.float32)
            feat_dict[patno] = {"feat": feat, "available": 1.0}
        else:
            feat_dict[patno] = {
                "feat": np.full(MRI_FEAT_DIM, MRI_MISSING_FILL, dtype=np.float32),
                "available": 0.0
            }

    print("✅ MRI patients loaded:", len(feat_dict))
    return feat_dict, feat_cols

mri_path = find_mri_feature_file()
mri_feat_dict, MRI_FEATURE_COLUMNS = load_preextracted_mri_features(mri_path)
print("✅ Clinical raw dim:", CLINICAL_RAW_DIM)
print("✅ MRI feature dim:", len(MRI_FEATURE_COLUMNS))


# ==========================================================
# 9) SEQUENCES WITH MRI/IMAGE LATE-FUSION INPUT (ALLOW NaN)
# ==========================================================
def make_sequences_with_y_allow_nan(df_seq, months, feature_cols, label_col, last_month):
    X_list, y_list, pat_list = [], [], []

    for patno, g in df_seq.groupby("PATNO"):
        g = g.sort_values("MONTH").drop_duplicates(["MONTH"], keep="last").set_index("MONTH")

        if last_month not in g.index:
            continue

        y = g.loc[last_month, label_col]
        if pd.isna(y):
            continue

        gg = g.reindex(months)

        # Clinical longitudinal features.
        Xc = gg[feature_cols].to_numpy(dtype=np.float32)

        # MRI/image features are patient-level and repeated at every time point.
        if int(patno) in mri_feat_dict:
            xmri = mri_feat_dict[int(patno)]["feat"].astype(np.float32)
            xflag = np.array([mri_feat_dict[int(patno)]["available"]], dtype=np.float32)
        else:
            xmri = np.full(MRI_FEAT_DIM, MRI_MISSING_FILL, dtype=np.float32)
            xflag = np.array([0.0], dtype=np.float32)

        T = Xc.shape[0]
        xmri_rep = np.repeat(xmri.reshape(1, -1), T, axis=0)
        xflag_rep = np.repeat(xflag.reshape(1, -1), T, axis=0)

        # Raw fused input = [clinical time-series | MRI/image features | MRI available flag]
        Xfused = np.concatenate([Xc, xmri_rep, xflag_rep], axis=1).astype(np.float32)

        X_list.append(Xfused)
        y_list.append(int(y))
        pat_list.append(int(patno))

    if len(X_list) == 0:
        raise ValueError(f"Aucune séquence construite (label={label_col}, last_month={last_month}).")

    return (
        np.stack(X_list).astype(np.float32),
        np.array(y_list, dtype=np.int64),
        np.array(pat_list, dtype=np.int64)
    )

months_2 = [0, 12]
months_4 = [0, 12, 24, 36]
months_6 = [0, 12, 24, 36, 48, 60]

X2, y2, p2 = make_sequences_with_y_allow_nan(df_seq, months_2, dep_cols, "GDS__DEPRESSION", 12)
X4, y4, p4 = make_sequences_with_y_allow_nan(df_seq, months_4, dep_cols, "GDS__DEPRESSION", 36)
X6, y6, p6 = make_sequences_with_y_allow_nan(df_seq, months_6, dep_cols, "GDS__DEPRESSION", 60)

print("✅ T2 fused:", X2.shape, "unique patients:", len(np.unique(p2)))
print("✅ T4 fused:", X4.shape, "unique patients:", len(np.unique(p4)))
print("✅ T6 fused:", X6.shape, "unique patients:", len(np.unique(p6)))

np.save(OUT_DIR / "X2_dep_mri_fused.npy", X2)
np.save(OUT_DIR / "y2_dep_mri_fused.npy", y2)
np.save(OUT_DIR / "pat2_dep_mri_fused.npy", p2)

np.save(OUT_DIR / "X4_dep_mri_fused.npy", X4)
np.save(OUT_DIR / "y4_dep_mri_fused.npy", y4)
np.save(OUT_DIR / "pat4_dep_mri_fused.npy", p4)

np.save(OUT_DIR / "X6_dep_mri_fused.npy", X6)
np.save(OUT_DIR / "y6_dep_mri_fused.npy", y6)
np.save(OUT_DIR / "pat6_dep_mri_fused.npy", p6)


# ==========================================================
# 10) SPLIT PATIENT-WISE 70/15/15
# ==========================================================
def split_train_val_test_patientwise(
    X, y, pat,
    train_ratio=0.70,
    val_ratio=0.15,
    test_ratio=0.15,
    seed=42
):
    assert abs(train_ratio + val_ratio + test_ratio - 1.0) < 1e-9

    unique_pat = np.unique(pat)

    gss1 = GroupShuffleSplit(
        n_splits=1,
        test_size=(val_ratio + test_ratio),
        random_state=seed
    )
    tr_pat_idx, tmp_pat_idx = next(gss1.split(unique_pat, groups=unique_pat))
    pat_tr_set = set(unique_pat[tr_pat_idx])
    pat_tmp = unique_pat[tmp_pat_idx]

    test_share_of_temp = test_ratio / (val_ratio + test_ratio)
    gss2 = GroupShuffleSplit(
        n_splits=1,
        test_size=test_share_of_temp,
        random_state=seed
    )
    va_pat_idx, te_pat_idx = next(gss2.split(pat_tmp, groups=pat_tmp))
    pat_va_set = set(pat_tmp[va_pat_idx])
    pat_te_set = set(pat_tmp[te_pat_idx])

    tr_idx = np.array([i for i, pp in enumerate(pat) if pp in pat_tr_set], dtype=int)
    va_idx = np.array([i for i, pp in enumerate(pat) if pp in pat_va_set], dtype=int)
    te_idx = np.array([i for i, pp in enumerate(pat) if pp in pat_te_set], dtype=int)

    return (
        (X[tr_idx], y[tr_idx], pat[tr_idx]),
        (X[va_idx], y[va_idx], pat[va_idx]),
        (X[te_idx], y[te_idx], pat[te_idx]),
    )

def print_split_info(name, tr, va, te):
    X_tr, y_tr, pat_tr = tr
    X_va, y_va, pat_va = va
    X_te, y_te, pat_te = te

    print(f"\n{name} SPLIT:")
    print("Train:", X_tr.shape, "patients:", len(np.unique(pat_tr)), "class:", np.unique(y_tr, return_counts=True))
    print("Val  :", X_va.shape, "patients:", len(np.unique(pat_va)), "class:", np.unique(y_va, return_counts=True))
    print("Test :", X_te.shape, "patients:", len(np.unique(pat_te)), "class:", np.unique(y_te, return_counts=True))
    print(
        "Train∩Val:", len(set(pat_tr) & set(pat_va)),
        "Train∩Test:", len(set(pat_tr) & set(pat_te)),
        "Val∩Test:", len(set(pat_va) & set(pat_te))
    )

t2_tr, t2_va, t2_te = split_train_val_test_patientwise(X2, y2, p2, TRAIN_RATIO, VAL_RATIO, TEST_RATIO, seed=42)
t4_tr, t4_va, t4_te = split_train_val_test_patientwise(X4, y4, p4, TRAIN_RATIO, VAL_RATIO, TEST_RATIO, seed=42)
t6_tr, t6_va, t6_te = split_train_val_test_patientwise(X6, y6, p6, TRAIN_RATIO, VAL_RATIO, TEST_RATIO, seed=42)

print_split_info("T2", t2_tr, t2_va, t2_te)
print_split_info("T4", t4_tr, t4_va, t4_te)
print_split_info("T6", t6_tr, t6_va, t6_te)


# ==========================================================
# 11) PREPROCESSING TRAIN ONLY + NaN MASK
# ==========================================================
def fit_preprocess_train_only_with_mask(X_train):
    X_train = X_train.astype(np.float32)
    N, T, F = X_train.shape

    mask = np.isfinite(X_train).astype(np.float32)

    flat = X_train.reshape(-1, F)
    med = np.nanmedian(flat, axis=0)
    med = np.where(np.isfinite(med), med, 0.0)

    flat_imp = flat.copy()
    inds = np.where(~np.isfinite(flat_imp))
    flat_imp[inds] = np.take(med, inds[1])

    scaler = MinMaxScaler()
    flat_scaled = scaler.fit_transform(flat_imp)
    X_scaled = flat_scaled.reshape(N, T, F).astype(np.float32)

    X_out = np.concatenate([X_scaled, mask], axis=-1).astype(np.float32)
    return X_out, med.astype(np.float32), scaler

def transform_with_train_params_with_mask(X, med, scaler):
    X = X.astype(np.float32)
    N, T, F = X.shape

    mask = np.isfinite(X).astype(np.float32)

    flat = X.reshape(-1, F)
    flat_imp = flat.copy()
    inds = np.where(~np.isfinite(flat_imp))
    flat_imp[inds] = np.take(med, inds[1])

    flat_scaled = scaler.transform(flat_imp)
    X_scaled = flat_scaled.reshape(N, T, F).astype(np.float32)

    X_out = np.concatenate([X_scaled, mask], axis=-1).astype(np.float32)
    return X_out


# ==========================================================
# 12) MODEL: HYBRID TRANSFORMER + GRU WITH FINAL MRI/IMAGE FUSION
# ==========================================================
class HybridTransformerGRU(nn.Module):
    """
    Late fusion:
      1) clinical longitudinal variables go through Transformer + BiGRU
      2) MRI/image features are added only at the final fusion head
    Input after preprocessing has layout:
      [scaled clinical | scaled MRI | scaled MRI_flag | mask clinical | mask MRI | mask MRI_flag]
    """
    def __init__(
        self,
        input_dim,
        max_len=6,
        d_model=128,
        nhead=4,
        num_layers=2,
        dim_feedforward=256,
        gru_hidden=64,
        gru_layers=1,
        dropout=0.30,
        clinical_raw_dim=None,
        mri_feat_dim=MRI_FEAT_DIM,
        flag_dim=1
    ):
        super().__init__()

        if clinical_raw_dim is None:
            clinical_raw_dim = CLINICAL_RAW_DIM

        if d_model % nhead != 0:
            raise ValueError(f"d_model ({d_model}) must be divisible by nhead ({nhead})")

        self.clinical_raw_dim = int(clinical_raw_dim)
        self.mri_feat_dim = int(mri_feat_dim)
        self.flag_dim = int(flag_dim)
        self.raw_total_dim = self.clinical_raw_dim + self.mri_feat_dim + self.flag_dim

        clinical_input_dim = self.clinical_raw_dim * 2
        mri_fusion_dim = (self.mri_feat_dim + self.flag_dim) * 2

        self.input_proj = nn.Sequential(
            nn.Linear(clinical_input_dim, d_model),
            nn.LayerNorm(d_model),
            nn.GELU(),
            nn.Dropout(dropout)
        )
        self.pos = nn.Parameter(torch.randn(1, max_len, d_model) * 0.02)

        enc_layer = nn.TransformerEncoderLayer(
            d_model=d_model,
            nhead=nhead,
            dim_feedforward=dim_feedforward,
            dropout=dropout,
            batch_first=True,
            activation="gelu",
            norm_first=False
        )
        self.encoder = nn.TransformerEncoder(enc_layer, num_layers=num_layers)
        self.tr_norm = nn.LayerNorm(d_model)

        self.tr_attn = nn.Sequential(
            nn.Linear(d_model, d_model // 2),
            nn.Tanh(),
            nn.Dropout(dropout),
            nn.Linear(d_model // 2, 1)
        )

        self.gru = nn.GRU(
            input_size=clinical_input_dim,
            hidden_size=gru_hidden,
            num_layers=gru_layers,
            batch_first=True,
            dropout=(dropout if gru_layers > 1 else 0.0),
            bidirectional=True
        )
        self.gru_norm = nn.LayerNorm(gru_hidden * 2)

        clinical_context_dim = d_model + (gru_hidden * 2)
        self.mri_branch = nn.Sequential(
            nn.Linear(mri_fusion_dim, 128),
            nn.LayerNorm(128),
            nn.GELU(),
            nn.Dropout(dropout),
            nn.Linear(128, 64),
            nn.GELU(),
        )

        fusion_dim = clinical_context_dim + 64
        self.head = nn.Sequential(
            nn.Linear(fusion_dim, 128),
            nn.LayerNorm(128),
            nn.GELU(),
            nn.Dropout(dropout),
            nn.Linear(128, 1)
        )

    def split_clinical_mri(self, x):
        C = self.clinical_raw_dim
        M = self.mri_feat_dim
        F = self.flag_dim
        R = self.raw_total_dim

        scaled = x[:, :, :R]
        masks = x[:, :, R:R+R]

        clinical_scaled = scaled[:, :, :C]
        clinical_mask = masks[:, :, :C]
        x_clinical = torch.cat([clinical_scaled, clinical_mask], dim=-1)

        mri_start = C
        mri_end = C + M + F
        mri_scaled_last = scaled[:, -1, mri_start:mri_end]
        mri_mask_last = masks[:, -1, mri_start:mri_end]
        x_mri = torch.cat([mri_scaled_last, mri_mask_last], dim=-1)

        return x_clinical, x_mri

    def forward(self, x):
        x_clinical, x_mri = self.split_clinical_mri(x)

        h = self.input_proj(x_clinical)
        h = h + self.pos[:, :x.size(1), :]
        h = self.encoder(h)
        h = self.tr_norm(h)

        tr_scores = self.tr_attn(h).squeeze(-1)
        tr_w = torch.softmax(tr_scores, dim=1)
        tr_ctx = torch.sum(h * tr_w.unsqueeze(-1), dim=1)

        g, _ = self.gru(x_clinical)
        g_last = self.gru_norm(g[:, -1, :])

        mri_ctx = self.mri_branch(x_mri)

        z = torch.cat([tr_ctx, g_last, mri_ctx], dim=1)
        logits = self.head(z).squeeze(-1)
        return logits


# ==========================================================
# 13) HELPERS
# ==========================================================
def find_best_threshold_f1_pos(y_true, probs):
    best_thr = 0.5
    best_f1 = -1.0
    best_bacc = -1.0

    for thr in np.linspace(0.10, 0.90, 81):
        pred = (probs >= thr).astype(int)
        f1 = f1_score(y_true, pred, zero_division=0)
        bacc = balanced_accuracy_score(y_true, pred)

        if (f1 > best_f1) or (np.isclose(f1, best_f1) and bacc > best_bacc):
            best_f1 = f1
            best_bacc = bacc
            best_thr = thr

    return float(best_thr), float(best_f1), float(best_bacc)

def smooth_labels_torch(y, smoothing=0.03):
    if smoothing <= 0:
        return y
    return y * (1.0 - smoothing) + 0.5 * smoothing

def augment_noise(X, scale=0.01):
    if scale <= 0:
        return X.astype(np.float32)
    noise = np.random.normal(0, scale, X.shape).astype(np.float32)
    return (X + noise).astype(np.float32)

def apply_input_dropout(X, p=0.05, fill_value=0.0):
    if p is None or p <= 0:
        return X.astype(np.float32)
    X = X.astype(np.float32, copy=True)
    mask = (np.random.rand(*X.shape) < p)
    X[mask] = np.float32(fill_value)
    return X

class FocalLossWithLogits(nn.Module):
    def __init__(self, gamma=2.0, alpha=0.25, pos_weight=None, reduction="mean"):
        super().__init__()
        self.gamma = float(gamma)
        self.alpha = float(alpha)
        self.pos_weight = pos_weight
        self.reduction = reduction

    def forward(self, logits, targets):
        bce = nn.functional.binary_cross_entropy_with_logits(
            logits, targets, pos_weight=self.pos_weight, reduction="none"
        )
        probs = torch.sigmoid(logits)
        pt = probs * targets + (1.0 - probs) * (1.0 - targets)
        focal = (1.0 - pt).pow(self.gamma)
        alpha_t = self.alpha * targets + (1.0 - self.alpha) * (1.0 - targets)
        loss = alpha_t * focal * bce

        if self.reduction == "sum":
            return loss.sum()
        if self.reduction == "none":
            return loss
        return loss.mean()

class EMA:
    def __init__(self, model, decay=0.995):
        self.decay = float(decay)
        self.shadow = {}
        for name, p in model.named_parameters():
            if p.requires_grad:
                self.shadow[name] = p.detach().clone()

    @torch.no_grad()
    def update(self, model):
        d = self.decay
        for name, p in model.named_parameters():
            if p.requires_grad:
                self.shadow[name].mul_(d).add_(p.detach(), alpha=(1.0 - d))

    @torch.no_grad()
    def copy_to(self, model):
        for name, p in model.named_parameters():
            if name in self.shadow:
                p.data.copy_(self.shadow[name].data)

def plot_losses(tr, va, title):
    plt.figure(figsize=(6, 4))
    plt.plot(tr, label="train_loss")
    plt.plot(va, label="val_loss")
    plt.title(title)
    plt.xlabel("epoch")
    plt.ylabel("loss")
    plt.legend()
    plt.tight_layout()
    plt.show()


# ==========================================================
# 14) HORIZON CONFIGS
# ==========================================================
HORIZON_DEFAULTS = {
    "T2": {
        "d_model": 64,
        "nhead": 4,
        "num_layers": 1,
        "dim_feedforward": 128,
        "gru_hidden": 32,
        "gru_layers": 1,
        "dropout": 0.20,
        "max_epochs": 60,
        "patience": 6,
        "label_smooth": 0.02,
        "aug_noise": 0.005,
        "input_drop_p": 0.02,
        "focal_gamma": 1.5,
        "focal_alpha": 0.35
    },
    "T4": {
        "d_model": 128,
        "nhead": 4,
        "num_layers": 2,
        "dim_feedforward": 256,
        "gru_hidden": 64,
        "gru_layers": 1,
        "dropout": 0.25,
        "max_epochs": 60,
        "patience": 8,
        "label_smooth": 0.03,
        "aug_noise": 0.01,
        "input_drop_p": 0.03,
        "focal_gamma": 2.0,
        "focal_alpha": 0.30
    },
    "T6": {
        "d_model": 128,
        "nhead": 4,
        "num_layers": 2,
        "dim_feedforward": 256,
        "gru_hidden": 64,
        "gru_layers": 1,
        "dropout": 0.30,
        "max_epochs": 60,
        "patience": 10,
        "label_smooth": 0.03,
        "aug_noise": 0.01,
        "input_drop_p": 0.02,
        "focal_gamma": 1.0,
        "focal_alpha": 0.40
    }
}


# ==========================================================
# 15) TRAIN HYBRID MODEL
# ==========================================================
def train_hybrid_model(
    Ztr, ytr, Zva, yva,
    d_model=128,
    nhead=4,
    num_layers=2,
    dim_feedforward=256,
    gru_hidden=64,
    gru_layers=1,
    dropout=0.30,
    lr=5e-4,
    weight_decay=1e-3,
    batch_size=64,
    max_epochs=60,
    patience=8,
    label_smooth=0.03,
    grad_clip=1.0,
    aug_noise=0.01,
    input_drop_p=0.05,
    ema_decay=0.995,
    focal_gamma=2.0,
    focal_alpha=0.25,
    max_len=6,
    device=None
):
    if device is None:
        device = torch.device("cpu")

    Ztr = Ztr.astype(np.float32, copy=False)
    Zva = Zva.astype(np.float32, copy=False)
    ytr = ytr.astype(np.float32, copy=False)
    yva = yva.astype(np.float32, copy=False)

    Ztr_aug = augment_noise(Ztr, scale=aug_noise)
    Ztr_aug = apply_input_dropout(Ztr_aug, p=input_drop_p, fill_value=0.0)

    pos = float((ytr == 1).sum())
    neg = float((ytr == 0).sum())
    pos_weight = torch.tensor([neg / max(pos, 1.0)], device=device, dtype=torch.float32)

    criterion = FocalLossWithLogits(
        gamma=focal_gamma,
        alpha=focal_alpha,
        pos_weight=pos_weight
    )

    model = HybridTransformerGRU(
        input_dim=Ztr.shape[-1],
        max_len=max_len,
        d_model=d_model,
        nhead=nhead,
        num_layers=num_layers,
        dim_feedforward=dim_feedforward,
        gru_hidden=gru_hidden,
        gru_layers=gru_layers,
        dropout=dropout
    ).to(device)

    optimizer = torch.optim.AdamW(model.parameters(), lr=lr, weight_decay=weight_decay)
    scheduler = torch.optim.lr_scheduler.ReduceLROnPlateau(
        optimizer, mode="min", factor=0.5, patience=2
    )
    ema = EMA(model, decay=ema_decay)

    xtr_tensor = torch.tensor(Ztr_aug, dtype=torch.float32)
    ytr_tensor = torch.tensor(ytr, dtype=torch.float32)
    xva_tensor = torch.tensor(Zva, dtype=torch.float32)
    yva_tensor = torch.tensor(yva, dtype=torch.float32)

    train_ds = TensorDataset(xtr_tensor, ytr_tensor)
    val_ds = TensorDataset(xva_tensor, yva_tensor)

    train_dl = DataLoader(
        train_ds,
        batch_size=batch_size,
        shuffle=True,
        num_workers=NUM_WORKERS,
        pin_memory=PIN_MEMORY,
        persistent_workers=(NUM_WORKERS > 0)
    )

    val_dl = DataLoader(
        val_ds,
        batch_size=max(128, batch_size * 2),
        shuffle=False,
        num_workers=NUM_WORKERS,
        pin_memory=PIN_MEMORY,
        persistent_workers=(NUM_WORKERS > 0)
    )

    best_val = float("inf")
    best_state = None
    best_ema_state = None
    wait = 0

    tr_losses, va_losses = [], []

    for ep in range(1, max_epochs + 1):
        model.train()
        total = 0.0

        for xb, yb in train_dl:
            xb = xb.to(device)
            yb = yb.to(device)

            if label_smooth > 0:
                yb = smooth_labels_torch(yb, smoothing=label_smooth)

            optimizer.zero_grad(set_to_none=True)
            logits = model(xb)
            loss = criterion(logits, yb)
            loss.backward()

            if grad_clip is not None and grad_clip > 0:
                torch.nn.utils.clip_grad_norm_(model.parameters(), grad_clip)

            optimizer.step()
            ema.update(model)
            total += loss.item()

        train_loss = total / max(len(train_dl), 1)

        model.eval()
        backup = {k: v.detach().cpu().clone() for k, v in model.state_dict().items()}
        ema.copy_to(model)

        val_total = 0.0
        probs_list = []

        with torch.no_grad():
            for xb, yb in val_dl:
                xb = xb.to(device)
                yb = yb.to(device)

                logits = model(xb)
                loss = criterion(logits, yb)

                val_total += loss.item()
                probs_list.append(torch.sigmoid(logits).detach().cpu())

        val_loss = val_total / max(len(val_dl), 1)
        probs = torch.cat(probs_list).numpy().reshape(-1)
        thr, f1, bacc = find_best_threshold_f1_pos(yva.astype(int), probs)

        model.load_state_dict(backup)

        tr_losses.append(train_loss)
        va_losses.append(val_loss)
        scheduler.step(val_loss)

        if ep == 1 or ep % 5 == 0:
            lr_now = optimizer.param_groups[0]["lr"]
            print(
                f"[HybridTransformerGRU] ep{ep:02d} lr={lr_now:.2e} "
                f"train_loss={train_loss:.5f} val_loss={val_loss:.5f} "
                f"val_bacc={bacc:.4f} val_f1={f1:.4f} thr={thr:.2f}"
            )

        if val_loss < best_val:
            best_val = val_loss
            best_state = {k: v.detach().cpu().clone() for k, v in model.state_dict().items()}
            best_ema_state = {k: v.detach().cpu().clone() for k, v in ema.shadow.items()}
            wait = 0
        else:
            wait += 1
            if wait >= patience:
                print("✅ Early stopping triggered.")
                break

    if best_state is not None:
        model.load_state_dict(best_state)

    if best_ema_state is not None:
        with torch.no_grad():
            for name, p in model.named_parameters():
                if name in best_ema_state:
                    p.data.copy_(best_ema_state[name].data)

    model.to(device).eval()

    probs_list = []
    with torch.no_grad():
        for xb, _ in val_dl:
            xb = xb.to(device)
            logits = model(xb)
            probs_list.append(torch.sigmoid(logits).detach().cpu())

    probs = torch.cat(probs_list).numpy().reshape(-1)
    best_thr, best_f1, _ = find_best_threshold_f1_pos(yva.astype(int), probs)

    return model, tr_losses, va_losses, best_thr, best_f1


# ==========================================================
# 16) (FOLD SUPPRIMÉ) - on utilise uniquement le split train/val basé sur le seed
# ==========================================================
# ❌ GroupKFold retiré : plus de validation croisée multi-folds.
# ✅ Optuna utilise directement (Z_tr, y_tr) / (Z_va, y_va) du split patient-wise par seed.


# ==========================================================
# 17) OPTUNA OBJECTIVE
# ==========================================================
def optuna_objective(trial, Ztr, ytr, Zva, yva, horizon_name="T2"):
    defaults = HORIZON_DEFAULTS[horizon_name]

    if defaults["d_model"] >= 128:
        d_model_choices = [defaults["d_model"], defaults["d_model"] + 64]
    else:
        d_model_choices = [64, 128]

    d_model = trial.suggest_categorical("d_model", d_model_choices)

    possible_heads = [2, 4, 8]
    valid_heads = [h for h in possible_heads if d_model % h == 0]
    nhead = trial.suggest_categorical("nhead", valid_heads)

    num_layers = trial.suggest_int(
        "num_layers",
        max(1, defaults["num_layers"] - 1),
        defaults["num_layers"] + 1
    )

    dim_feedforward = trial.suggest_categorical(
        "dim_feedforward",
        sorted(set([
            defaults["dim_feedforward"],
            defaults["dim_feedforward"] + 128
        ]))
    )

    gru_hidden = trial.suggest_categorical(
        "gru_hidden",
        sorted(set([32, defaults["gru_hidden"], 96]))
    )

    gru_layers = trial.suggest_int("gru_layers", 1, 2)

    dropout = trial.suggest_float(
        "dropout",
        max(0.10, defaults["dropout"] - 0.10),
        min(0.50, defaults["dropout"] + 0.10)
    )

    lr = trial.suggest_float("lr", 1e-4, 1e-3, log=True)
    weight_decay = trial.suggest_float("weight_decay", 1e-5, 3e-3, log=True)

    aug_noise = trial.suggest_float(
        "aug_noise",
        max(0.0, defaults["aug_noise"] - 0.005),
        min(0.03, defaults["aug_noise"] + 0.01)
    )

    input_drop_p = trial.suggest_float(
        "input_drop_p",
        max(0.0, defaults["input_drop_p"] - 0.02),
        min(0.10, defaults["input_drop_p"] + 0.03)
    )

    focal_gamma = trial.suggest_float(
        "focal_gamma",
        max(0.5, defaults["focal_gamma"] - 0.7),
        min(3.0, defaults["focal_gamma"] + 0.7)
    )

    focal_alpha = trial.suggest_float(
        "focal_alpha",
        max(0.15, defaults["focal_alpha"] - 0.10),
        min(0.50, defaults["focal_alpha"] + 0.10)
    )

    search_epochs = min(25, defaults["max_epochs"])
    search_patience = min(5, defaults["patience"])

    # ✅ Pas de fold : un seul entraînement sur (Ztr, ytr) / (Zva, yva) du split du seed courant
    _, _, _, _, f1 = train_hybrid_model(
        Ztr, ytr, Zva, yva,
        d_model=d_model,
        nhead=nhead,
        num_layers=num_layers,
        dim_feedforward=dim_feedforward,
        gru_hidden=gru_hidden,
        gru_layers=gru_layers,
        dropout=dropout,
        lr=lr,
        weight_decay=weight_decay,
        aug_noise=aug_noise,
        input_drop_p=input_drop_p,
        batch_size=64,
        max_epochs=search_epochs,
        patience=search_patience,
        label_smooth=defaults["label_smooth"],
        grad_clip=1.0,
        focal_gamma=focal_gamma,
        focal_alpha=focal_alpha,
        max_len=Ztr.shape[1],
        device=device
    )

    trial.report(float(f1), step=0)
    if trial.should_prune():
        raise optuna.exceptions.TrialPruned()

    return float(f1)


# ==========================================================
# 18) RUN ONE HORIZON (MODIFIED TO RETURN METRICS)
# ==========================================================
def run_one_horizon_hybrid(X, y, pat, name="T2", n_trials=20, seed=42):
    print("\n" + "=" * 70)
    print(f"RUN HYBRID TRANSFORMER + GRU - {name}")
    print("=" * 70)
    
    # Set seed for reproducibility
    set_seed(seed)

    defaults = HORIZON_DEFAULTS[name]

    (X_tr, y_tr, pat_tr), (X_va, y_va, pat_va), (X_te, y_te, pat_te) = split_train_val_test_patientwise(
        X, y, pat,
        train_ratio=TRAIN_RATIO,
        val_ratio=VAL_RATIO,
        test_ratio=TEST_RATIO,
        seed=seed
    )

    print("Shapes:")
    print("Train:", X_tr.shape, "patients:", len(np.unique(pat_tr)))
    print("Val  :", X_va.shape, "patients:", len(np.unique(pat_va)))
    print("Test :", X_te.shape, "patients:", len(np.unique(pat_te)))

    X_tr_pp, med, scaler = fit_preprocess_train_only_with_mask(X_tr)
    X_va_pp = transform_with_train_params_with_mask(X_va, med, scaler)
    X_te_pp = transform_with_train_params_with_mask(X_te, med, scaler)

    print("✅ After preprocessing + mask:")
    print("Train:", X_tr_pp.shape, "Val:", X_va_pp.shape, "Test:", X_te_pp.shape)

    Z_tr = X_tr_pp
    Z_va = X_va_pp
    Z_te = X_te_pp

    # ✅ Pas de fold : Optuna optimise directement sur le split train/val du seed courant
    print(f"✅ Optuna sans fold : Ztr={Z_tr.shape} Zva={Z_va.shape}")

    sampler = TPESampler(seed=seed)
    pruner = MedianPruner(n_startup_trials=5)

    study = optuna.create_study(
        direction="maximize",
        sampler=sampler,
        pruner=pruner
    )

    study.optimize(
        lambda trial: optuna_objective(trial, Z_tr, y_tr, Z_va, y_va, horizon_name=name),
        n_trials=n_trials,
        show_progress_bar=True
    )

    best = study.best_params
    print("\n✅ BEST PARAMS:", best)

    final_batch_size = 64

    model, tr_loss, va_loss, best_thr_val, best_f1_val = train_hybrid_model(
        Z_tr, y_tr,
        Z_va, y_va,
        d_model=best["d_model"],
        nhead=best["nhead"],
        num_layers=best["num_layers"],
        dim_feedforward=best["dim_feedforward"],
        gru_hidden=best["gru_hidden"],
        gru_layers=best["gru_layers"],
        dropout=best["dropout"],
        lr=best["lr"],
        weight_decay=best["weight_decay"],
        aug_noise=best["aug_noise"],
        input_drop_p=best["input_drop_p"],
        batch_size=final_batch_size,
        max_epochs=defaults["max_epochs"],
        patience=defaults["patience"],
        label_smooth=defaults["label_smooth"],
        grad_clip=1.0,
        focal_gamma=best["focal_gamma"],
        focal_alpha=best["focal_alpha"],
        max_len=Z_tr.shape[1],
        device=device
    )

    plot_losses(tr_loss, va_loss, f"{name} - Hybrid Transformer+GRU train/val loss")

    model.eval()
    test_ds = TensorDataset(torch.tensor(Z_te, dtype=torch.float32))
    test_dl = DataLoader(
        test_ds,
        batch_size=max(128, final_batch_size),
        shuffle=False,
        num_workers=NUM_WORKERS,
        pin_memory=PIN_MEMORY,
        persistent_workers=(NUM_WORKERS > 0)
    )

    probs_list = []
    with torch.no_grad():
        for (xb,) in test_dl:
            xb = xb.to(device)
            logits = model(xb)
            probs_list.append(torch.sigmoid(logits).detach().cpu())

    probs = torch.cat(probs_list).numpy().reshape(-1)
    pred = (probs >= best_thr_val).astype(int)

    # ======================================================
    # TEST METRICS
    # ======================================================
    test_acc = accuracy_score(y_te, pred)
    test_precision = precision_score(y_te, pred, average="macro", zero_division=0)
    test_recall = recall_score(y_te, pred, average="macro", zero_division=0)
    test_f1 = f1_score(y_te, pred, average="macro", zero_division=0)
    test_bacc = balanced_accuracy_score(y_te, pred)
    cm = confusion_matrix(y_te, pred, labels=[0, 1])
    per_class_report = classification_report(
        y_te, pred,
        labels=[0, 1],
        target_names=["No Depression", "Depression"],
        output_dict=True,
        zero_division=0
    )

    print("\n=== TEST REPORT ===")
    print(f"Accuracy: {test_acc:.4f}")
    print(f"Macro Precision: {test_precision:.4f}")
    print(f"Macro Recall: {test_recall:.4f}")
    print(f"Macro F1: {test_f1:.4f}")
    print(f"Balanced Accuracy: {test_bacc:.4f}")
    print("\nConfusion Matrix:")
    print(cm)
    print(classification_report(y_te, pred, labels=[0, 1], target_names=["No Depression", "Depression"], zero_division=0))

    # ======================================================
    # RETURN RESULTS
    # ======================================================
    return {
        "name": name,
        "seed": seed,
        "accuracy": float(test_acc),
        "precision": float(test_precision),
        "recall": float(test_recall),
        "f1": float(test_f1),
        "balanced_accuracy": float(test_bacc),
        "confusion_matrix": cm.tolist(),
        "per_class": per_class_report,
        "best_params": best,
        "best_thr_from_val": float(best_thr_val),
        "n_pat_train": int(len(np.unique(pat_tr))),
        "n_pat_val": int(len(np.unique(pat_va))),
        "n_pat_test": int(len(np.unique(pat_te))),
        "n_seq_train": int(X_tr.shape[0]),
        "n_seq_val": int(X_va.shape[0]),
        "n_seq_test": int(X_te.shape[0]),
        # ✅ Courbes de loss (pour Mean Train/Val Loss sur les 5 seeds)
        "train_loss_curve": [float(v) for v in tr_loss],
        "val_loss_curve": [float(v) for v in va_loss],
        "final_train_loss": float(tr_loss[-1]) if len(tr_loss) else float("nan"),
        "final_val_loss": float(va_loss[-1]) if len(va_loss) else float("nan"),
    }


# ==========================================================
# 19) RUN ALL WITH MULTI-SEED
# ==========================================================

SEEDS = [42, 43, 44, 45, 46]

all_results = {}

for horizon_name, (X, y, p) in {
    "T2": (X2, y2, p2),
    "T4": (X4, y4, p4),
    "T6": (X6, y6, p6)
}.items():

    print("\n" + "=" * 80)
    print(f"RUNNING {horizon_name}")
    print("=" * 80)

    acc_scores = []
    precision_scores = []
    recall_scores = []
    f1_scores = []
    bacc_scores = []
    confusion_matrices = []
    per_class_seed_results = []
    all_seed_results = []
    train_loss_curves_seeds = []
    val_loss_curves_seeds = []
    final_train_losses = []
    final_val_losses = []

    for seed in SEEDS:
        print("\n" + "-" * 60)
        print(f"SEED = {seed}")
        print("-" * 60)

        # run model with current seed
        res = run_one_horizon_hybrid(X, y, p, name=horizon_name, n_trials=N_TRIALS, seed=seed)
        
        print(f"\nTEST RESULTS FOR SEED = {seed}:")
        print(f"Accuracy = {res['accuracy']:.4f}")
        print(f"Macro Precision = {res['precision']:.4f}")
        print(f"Macro Recall = {res['recall']:.4f}")
        print(f"Macro F1 = {res['f1']:.4f}")
        print(f"Balanced Accuracy = {res['balanced_accuracy']:.4f}")
        print("Confusion Matrix:")
        print(np.array(res["confusion_matrix"]))

        # Store metrics
        acc_scores.append(res["accuracy"])
        precision_scores.append(res["precision"])
        recall_scores.append(res["recall"])
        f1_scores.append(res["f1"])
        bacc_scores.append(res["balanced_accuracy"])
        confusion_matrices.append(np.array(res["confusion_matrix"]))
        per_class_seed_results.append(res["per_class"])
        all_seed_results.append(res)

        # ✅ Loss curves (pour Mean Train/Val Loss sur les 5 seeds)
        train_loss_curves_seeds.append(res["train_loss_curve"])
        val_loss_curves_seeds.append(res["val_loss_curve"])
        final_train_losses.append(res["final_train_loss"])
        final_val_losses.append(res["final_val_loss"])

    # ======================================================
    # CALCUL MOYENNES ± STD
    # ======================================================

    def mean_std(x):
        if len(x) == 0:
            return 0, 0
        return np.mean(x), np.std(x)

    mean_acc, std_acc = mean_std(acc_scores)
    mean_prec, std_prec = mean_std(precision_scores)
    mean_rec, std_rec = mean_std(recall_scores)
    mean_f1, std_f1 = mean_std(f1_scores)
    mean_bacc, std_bacc = mean_std(bacc_scores)

    print("\n" + "=" * 80)
    print(f"AVERAGE RESULTS - {horizon_name}")
    print("=" * 80)
    print(f"Accuracy: {mean_acc:.4f} ± {std_acc:.4f}")
    print(f"Precision: {mean_prec:.4f} ± {std_prec:.4f}")
    print(f"Recall: {mean_rec:.4f} ± {std_rec:.4f}")
    print(f"F1-score: {mean_f1:.4f} ± {std_f1:.4f}")
    print(f"Balanced Accuracy: {mean_bacc:.4f} ± {std_bacc:.4f}")

    # ======================================================
    # PER-CLASS RESULTS MEAN ± STD ACROSS SEEDS
    # ======================================================
    class_names = ["No Depression", "Depression"]
    per_class_summary = {}

    print("\nPer-class results (mean ± std across seeds):")
    for cls_name in class_names:
        cls_prec = [r[cls_name]["precision"] for r in per_class_seed_results if cls_name in r]
        cls_rec = [r[cls_name]["recall"] for r in per_class_seed_results if cls_name in r]
        cls_f1 = [r[cls_name]["f1-score"] for r in per_class_seed_results if cls_name in r]
        cls_sup = [r[cls_name]["support"] for r in per_class_seed_results if cls_name in r]

        mp, sp = mean_std(cls_prec)
        mr, sr = mean_std(cls_rec)
        mf, sf = mean_std(cls_f1)
        ms, ss = mean_std(cls_sup)

        per_class_summary[cls_name] = {
            "precision": f"{mp:.4f} ± {sp:.4f}",
            "recall": f"{mr:.4f} ± {sr:.4f}",
            "f1-score": f"{mf:.4f} ± {sf:.4f}",
            "support": f"{ms:.1f} ± {ss:.1f}",
        }

        print(f"\n{cls_name}")
        print(f"  Precision: {mp:.4f} ± {sp:.4f}")
        print(f"  Recall   : {mr:.4f} ± {sr:.4f}")
        print(f"  F1-score : {mf:.4f} ± {sf:.4f}")
        print(f"  Support  : {ms:.1f} ± {ss:.1f}")

    # ======================================================
    # AVERAGE CONFUSION MATRIX
    # ======================================================
    if len(confusion_matrices) > 0:
        avg_cm = np.mean(np.stack(confusion_matrices), axis=0)
        std_cm = np.std(np.stack(confusion_matrices), axis=0)
        
        print("\nAverage Confusion Matrix (mean across seeds):")
        print(avg_cm)
        print("\nStd Confusion Matrix:")
        print(std_cm)

        # Plot average confusion matrix
        plt.figure(figsize=(6, 5))
        plt.imshow(avg_cm, cmap='Blues')
        plt.colorbar()
        plt.title(f"{horizon_name} - Mean Confusion Matrix (across seeds)")
        plt.xlabel("Predicted")
        plt.ylabel("True")
        
        for i in range(avg_cm.shape[0]):
            for j in range(avg_cm.shape[1]):
                plt.text(j, i, f"{avg_cm[i, j]:.1f}", ha="center", va="center", 
                        color="white" if avg_cm[i, j] > avg_cm.max()/2 else "black")
        plt.show()

    # ======================================================
    # MEAN TRAIN LOSS / MEAN VALIDATION LOSS SUR LES 5 SEEDS
    # ======================================================
    def pad_curve_to_length(curve, length):
        curve = list(curve)
        if len(curve) == 0:
            return [np.nan] * length
        if len(curve) >= length:
            return curve[:length]
        pad_value = curve[-1]
        return curve + [pad_value] * (length - len(curve))

    def mean_std_curves(curves):
        curves = [c for c in curves if len(c) > 0]
        if len(curves) == 0:
            return np.array([]), np.array([])
        max_len = max(len(c) for c in curves)
        padded = np.array([pad_curve_to_length(c, max_len) for c in curves], dtype=float)
        return np.nanmean(padded, axis=0), np.nanstd(padded, axis=0)

    mean_train_loss_curve, std_train_loss_curve = mean_std_curves(train_loss_curves_seeds)
    mean_val_loss_curve, std_val_loss_curve = mean_std_curves(val_loss_curves_seeds)

    mean_final_train_loss, std_final_train_loss = mean_std(final_train_losses)
    mean_final_val_loss, std_final_val_loss = mean_std(final_val_losses)

    print("\nMean Train Loss / Mean Validation Loss (5 seeds):")
    print(f"  Mean Train Loss (dernière epoch) : {mean_final_train_loss:.5f} ± {std_final_train_loss:.5f}")
    print(f"  Mean Val Loss   (dernière epoch) : {mean_final_val_loss:.5f} ± {std_final_val_loss:.5f}")

    # sauvegarde résultats
    all_results[horizon_name] = {
        "Accuracy": f"{mean_acc:.4f} ± {std_acc:.4f}",
        "Precision": f"{mean_prec:.4f} ± {std_prec:.4f}",
        "Recall": f"{mean_rec:.4f} ± {std_rec:.4f}",
        "F1-score": f"{mean_f1:.4f} ± {std_f1:.4f}",
        "Balanced Accuracy": f"{mean_bacc:.4f} ± {std_bacc:.4f}",
        "Per-class": per_class_summary,
        "Mean_confusion_matrix": avg_cm.tolist() if len(confusion_matrices) > 0 else [],
        "Std_confusion_matrix": std_cm.tolist() if len(confusion_matrices) > 0 else [],
        "Mean_Train_Loss_Final": f"{mean_final_train_loss:.5f} ± {std_final_train_loss:.5f}",
        "Mean_Val_Loss_Final": f"{mean_final_val_loss:.5f} ± {std_final_val_loss:.5f}",
        "Mean_Train_Loss_Curve": mean_train_loss_curve.tolist(),
        "Std_Train_Loss_Curve": std_train_loss_curve.tolist(),
        "Mean_Val_Loss_Curve": mean_val_loss_curve.tolist(),
        "Std_Val_Loss_Curve": std_val_loss_curve.tolist(),
        "Per_seed_results": all_seed_results
    }


# ==========================================================
# FINAL RESULTS
# ==========================================================

print("\n" + "=" * 80)
print("FINAL RESULTS")
print("=" * 80)

for horizon, metrics in all_results.items():
    print("\n" + "-" * 50)
    print(horizon)
    print("-" * 50)
    for metric, value in metrics.items():
        if metric not in ("Per_seed_results", "Mean_Train_Loss_Curve", "Std_Train_Loss_Curve",
                           "Mean_Val_Loss_Curve", "Std_Val_Loss_Curve",
                           "Mean_confusion_matrix", "Std_confusion_matrix"):
            print(f"{metric}: {value}")


# ==========================================================
# 20) FIGURE COMBINÉE - 3 MATRICES DE CONFUSION (T2 / T4 / T6)
# ==========================================================
HORIZON_ORDER = ["T2", "T4", "T6"]
HORIZON_ORDER = [h for h in HORIZON_ORDER if h in all_results]

fig_cm, axes_cm = plt.subplots(1, len(HORIZON_ORDER), figsize=(6 * len(HORIZON_ORDER), 5))
if len(HORIZON_ORDER) == 1:
    axes_cm = [axes_cm]

for ax, horizon_name in zip(axes_cm, HORIZON_ORDER):
    avg_cm = np.array(all_results[horizon_name]["Mean_confusion_matrix"])
    im = ax.imshow(avg_cm, cmap="Blues")
    ax.set_title(f"{horizon_name} - Mean Confusion Matrix (5 seeds)")
    ax.set_xlabel("Predicted")
    ax.set_ylabel("True")
    ax.set_xticks([0, 1])
    ax.set_xticklabels(["No Depression", "Depression"])
    ax.set_yticks([0, 1])
    ax.set_yticklabels(["No Depression", "Depression"])
    for i in range(avg_cm.shape[0]):
        for j in range(avg_cm.shape[1]):
            ax.text(
                j, i, f"{avg_cm[i, j]:.1f}",
                ha="center", va="center",
                color="white" if avg_cm[i, j] > avg_cm.max() / 2 else "black"
            )
    fig_cm.colorbar(im, ax=ax, fraction=0.046, pad=0.04)

fig_cm.suptitle("Matrices de confusion moyennes (mean across 5 seeds) - T2 / T4 / T6")
fig_cm.tight_layout()

cm_fig_path = OUT_DIR / "confusion_matrices_T2_T4_T6.png"
fig_cm.savefig(cm_fig_path, dpi=200, bbox_inches="tight")
plt.show()
print("✅ Figure des matrices de confusion sauvegardée :", cm_fig_path)


# ==========================================================
# 21) FIGURE COMBINÉE - 3 COURBES TRAIN LOSS / VAL LOSS (T2 / T4 / T6)
# ==========================================================
fig_loss, axes_loss = plt.subplots(1, len(HORIZON_ORDER), figsize=(6 * len(HORIZON_ORDER), 4.5))
if len(HORIZON_ORDER) == 1:
    axes_loss = [axes_loss]

for ax, horizon_name in zip(axes_loss, HORIZON_ORDER):
    mtr = np.array(all_results[horizon_name]["Mean_Train_Loss_Curve"])
    mva = np.array(all_results[horizon_name]["Mean_Val_Loss_Curve"])
    str_ = np.array(all_results[horizon_name]["Std_Train_Loss_Curve"])
    sva = np.array(all_results[horizon_name]["Std_Val_Loss_Curve"])

    if len(mtr) > 0:
        epochs = np.arange(1, len(mtr) + 1)
        ax.plot(epochs, mtr, label="Mean Train Loss", color="tab:blue")
        ax.fill_between(epochs, mtr - str_, mtr + str_, color="tab:blue", alpha=0.15)
        ax.plot(epochs, mva, label="Mean Val Loss", color="tab:orange")
        ax.fill_between(epochs, mva - sva, mva + sva, color="tab:orange", alpha=0.15)

    ax.set_title(f"{horizon_name} - Train/Val Loss (mean ± std, 5 seeds)")
    ax.set_xlabel("epoch")
    ax.set_ylabel("loss")
    ax.legend()

fig_loss.suptitle("Courbes Train Loss / Validation Loss moyennes (mean across 5 seeds) - T2 / T4 / T6")
fig_loss.tight_layout()

loss_fig_path = OUT_DIR / "train_val_loss_curves_T2_T4_T6.png"
fig_loss.savefig(loss_fig_path, dpi=200, bbox_inches="tight")
plt.show()
print("✅ Figure des courbes train/val loss sauvegardée :", loss_fig_path)


# ==========================================================
# 22) TABLEAU FINAL COMPLET DE TOUTES LES MÉTRIQUES
# ==========================================================
final_rows = []
for horizon_name in HORIZON_ORDER:
    m = all_results[horizon_name]
    row = {
        "Horizon": horizon_name,
        "Mean Accuracy ± Std": m["Accuracy"],
        "Mean Precision ± Std": m["Precision"],
        "Mean Recall ± Std": m["Recall"],
        "Mean F1 ± Std": m["F1-score"],
        "Mean Balanced Accuracy ± Std": m["Balanced Accuracy"],
        "Mean Train Loss ± Std": m["Mean_Train_Loss_Final"],
        "Mean Validation Loss ± Std": m["Mean_Val_Loss_Final"],
    }
    for cls_name in ["No Depression", "Depression"]:
        cls_metrics = m["Per-class"][cls_name]
        row[f"{cls_name} - Precision (mean ± std)"] = cls_metrics["precision"]
        row[f"{cls_name} - Recall (mean ± std)"] = cls_metrics["recall"]
        row[f"{cls_name} - F1-score (mean ± std)"] = cls_metrics["f1-score"]
        row[f"{cls_name} - Support (mean ± std)"] = cls_metrics["support"]
    final_rows.append(row)

final_metrics_df = pd.DataFrame(final_rows)

print("\n" + "=" * 80)
print("TABLEAU FINAL COMPLET DE TOUTES LES MÉTRIQUES")
print("=" * 80)
display(final_metrics_df)


# ==========================================================
# 23) EXPORTS AUTOMATIQUES : JSON / CSV / EXCEL
# ==========================================================

# ✅ Export JSON complet (toutes les métriques + résultats par seed + courbes de loss)
results_json_path = OUT_DIR / "final_results_multi_seed_mri_late_fusion.json"
with open(results_json_path, "w", encoding="utf-8") as f:
    json.dump(all_results, f, indent=2, ensure_ascii=False)
print("✅ Export JSON sauvegardé :", results_json_path)

# ✅ Export CSV court (résumé des métriques globales)
rows = []
for horizon, metrics in all_results.items():
    rows.append({
        "Horizon": horizon,
        "Accuracy": metrics["Accuracy"],
        "Precision": metrics["Precision"],
        "Recall": metrics["Recall"],
        "F1-score": metrics["F1-score"],
        "Balanced Accuracy": metrics["Balanced Accuracy"],
        "Mean Train Loss": metrics["Mean_Train_Loss_Final"],
        "Mean Validation Loss": metrics["Mean_Val_Loss_Final"],
    })
summary_df = pd.DataFrame(rows)
summary_csv_path = OUT_DIR / "summary_results_multi_seed_mri_late_fusion.csv"
summary_df.to_csv(summary_csv_path, index=False)
print("✅ Export CSV (résumé) sauvegardé :", summary_csv_path)

# ✅ Export CSV complet (avec per-class)
final_csv_path = OUT_DIR / "final_metrics_table_complete.csv"
final_metrics_df.to_csv(final_csv_path, index=False)
print("✅ Export CSV (complet, par classe) sauvegardé :", final_csv_path)

# ✅ Export Excel (résumé + complet + détail par seed pour chaque horizon)
final_xlsx_path = OUT_DIR / "final_metrics_table_complete.xlsx"
try:
    with pd.ExcelWriter(final_xlsx_path, engine="openpyxl") as writer:
        final_metrics_df.to_excel(writer, sheet_name="Resume_complet", index=False)
        summary_df.to_excel(writer, sheet_name="Resume_court", index=False)

        keep_cols = [
            "seed", "accuracy", "precision", "recall", "f1", "balanced_accuracy",
            "best_thr_from_val", "n_pat_train", "n_pat_val", "n_pat_test",
            "n_seq_train", "n_seq_val", "n_seq_test",
            "final_train_loss", "final_val_loss"
        ]
        for horizon_name in HORIZON_ORDER:
            per_seed_df = pd.DataFrame(all_results[horizon_name]["Per_seed_results"])
            cols_present = [c for c in keep_cols if c in per_seed_df.columns]
            per_seed_df[cols_present].to_excel(
                writer, sheet_name=f"{horizon_name}_par_seed", index=False
            )
    print("✅ Export Excel sauvegardé :", final_xlsx_path)
except Exception as e:
    print("⚠️ Export Excel impossible (openpyxl manquant ?) :", e)


# ==========================================================
# 24) AFFICHAGE AUTOMATIQUE FINAL
# ==========================================================
print("\n" + "=" * 80)
print("RÉCAPITULATIF DES FICHIERS EXPORTÉS")
print("=" * 80)
print("✅ JSON   :", results_json_path)
print("✅ CSV    :", summary_csv_path)
print("✅ CSV    :", final_csv_path)
print("✅ Excel  :", final_xlsx_path)
print("✅ PNG    :", cm_fig_path)
print("✅ PNG    :", loss_fig_path)

print("\n" + "=" * 80)
print("RÉSUMÉ COURT (Accuracy / Precision / Recall / F1 / Balanced Accuracy)")
print("=" * 80)
display(summary_df)

print("\n" + "=" * 80)
print("TABLEAU COMPLET (avec métriques par classe + loss)")
print("=" * 80)
display(final_metrics_df)
```
