```python
# ==========================================================
# COMPLETE FINAL CODE FROM START TO END
# MULTITASK PARALLEL TRANSFORMER / BiGRU + PRE-EXTRACTED MRI FEATURES
# 3-BRANCH LATE FUSION: tr_ctx (Transformer) + g_last (BiGRU) + mri_ctx (MRI)
# Transformer and BiGRU run IN PARALLEL on the same embedded sequence
# (NOT in series: BiGRU no longer consumes the Transformer output)
# T2 / T4 / T6 + OPTUNA (single split) + 5 SEEDS REPEATED HOLD-OUT (NO FOLDS)
# CPU VERSION - KAGGLE READY
# + MEAN/STD METRICS OVER 5 SEEDS (global + per-class)
# + MEAN CONFUSION MATRICES (rounded) PER HORIZON
# + FIGURE WITH THE 3 CONFUSION MATRICES (T2/T4/T6) PER TASK
# + FIGURE WITH THE 3 TRAIN/VAL LOSS CURVES (T2/T4/T6), averaged over seeds
# + FINAL METRICS TABLE + AUTOMATIC DISPLAY AT THE END
# ==========================================================

import re
import json
import random
import gc
import warnings
from contextlib import nullcontext
from pathlib import Path
from functools import reduce

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.model_selection import GroupShuffleSplit
from sklearn.preprocessing import MinMaxScaler, StandardScaler
from sklearn.metrics import (
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay,
    f1_score,
    balanced_accuracy_score,
    accuracy_score,
    precision_score,
    recall_score,
)

import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import DataLoader, TensorDataset, WeightedRandomSampler

import optuna
from optuna.samplers import TPESampler
from optuna.pruners import MedianPruner

try:
    from IPython.display import display
except Exception:
    def display(x):
        print(x)

warnings.filterwarnings("ignore")




# ==========================================================
# 0) CONFIG
# ==========================================================
OUT_DIR = Path("/kaggle/working/multitask_parallel_transformer_bigru_3branch_fusion")
OUT_DIR.mkdir(parents=True, exist_ok=True)

HORIZONS = {
    "T2": {"months": [0, 12], "last_month": 12},
    "T4": {"months": [0, 12, 24, 36], "last_month": 36},
    "T6": {"months": [0, 12, 24, 36, 48, 60], "last_month": 60},
}

SEED = 42                 # seed used for the Optuna hyperparameter search split
N_SEEDS = 5                # number of independent seeded runs (REPLACES the 5-fold CV, "sans fold")
SEEDS = [100, 200, 300, 400, 500]

TEST_RATIO = 0.15          # outer hold-out ratio (patient-level, GroupShuffleSplit) -> 15% TEST
INNER_VAL_RATIO = 0.17647058823529413  # inner train/val ratio -> 15% VAL of TOTAL (=> 70/15/15 overall)

SEARCH_EPOCHS = 20
FINAL_EPOCHS = 70
PATIENCE = 10
N_TRIALS = 18

NUM_WORKERS = 0
PIN_MEMORY = False

MRI_FEAT_DIM = 256
MRI_MISSING_FILL = -1.0

TASK_IMPORTANCE = {
    "subtype": 1.0,
    "depression": 1.0,
    "levels": 1.25,
}

MONTH_TO_INDEX = {0: 0, 12: 1, 24: 2, 36: 3, 48: 4, 60: 5}

CLASS_NAMES = {
    "subtype": ["TD", "PIGD"],
    "depression": ["NoDep", "Dep"],
    "levels": ["None", "Mild", "Moderate", "Severe"],
}
N_CLASSES = {"subtype": 2, "depression": 2, "levels": 4}




# ==========================================================
# 1) DEVICE
# ==========================================================
device = torch.device("cpu")
print("device:", device)




# ==========================================================
# 2) SEED
# ==========================================================
def set_seed(seed=42):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)

set_seed(SEED)


def autocast_context():
    return nullcontext()




# ==========================================================
# 3) FILE HELPERS
# ==========================================================
def find_file(filename):
    base = Path("/kaggle/input")
    matches = list(base.rglob(filename))
    if not matches:
        raise FileNotFoundError(f"{filename} introuvable dans /kaggle/input")
    print(f"[FOUND] {filename} -> {matches[0]}")
    return matches[0]


def find_mri_feature_file():
    candidates = list(Path("/kaggle/input").rglob("mri_featuress.csv"))
    if len(candidates) == 0:
        raise FileNotFoundError("mri_featuress.csv introuvable dans /kaggle/input")
    print("[FOUND] MRI feature file ->", candidates[0])
    return candidates[0]


def read_csv_robust(path):
    return pd.read_csv(
        path,
        encoding="latin1",
        sep=None,
        engine="python",
        on_bad_lines="skip"
    )




# ==========================================================
# 4) MERGE HELPERS
# ==========================================================
KEYS = ["PATNO", "EVENT_ID"]
DATE_CANDIDATES = ["INFODT", "VISITDT", "DATE", "EXAMDATE", "TESTDATE"]

ADMIN_WORDS = [
    "REC_ID", "PAG_NAME", "ORIG_ENTRY", "LAST_UPDATE",
    "ENTRY", "UPDATE", "NUPSOURC", "PTCGBOTH",
    "INFODT", "VISITDT", "EXAMDATE", "TESTDATE", "DATE"
]

LEAKY_CONTAINS = [
    "ESS_TOTAL", "SCORE_GDS", "SCAUTOT", "RBDTOT",
    "NP1RTOT", "NP2PTOT", "NP3TOT", "NHY"
]

_FR_MONTH = {
    "janv": "Jan", "févr": "Feb", "fevr": "Feb", "mars": "Mar", "avr": "Apr",
    "mai": "May", "juin": "Jun", "juil": "Jul", "août": "Aug", "aout": "Aug",
    "sept": "Sep", "oct": "Oct", "nov": "Nov", "déc": "Dec", "dec": "Dec"
}


def norm_cols(df):
    df = df.copy()
    df.columns = [str(c).strip().upper() for c in df.columns]
    return df


def coerce_keys(df):
    df = df.copy()
    if "PATNO" in df.columns:
        df["PATNO"] = pd.to_numeric(df["PATNO"], errors="coerce").astype("Int64")
    if "EVENT_ID" in df.columns:
        df["EVENT_ID"] = df["EVENT_ID"].astype(str).str.upper().str.strip()
    return df


def detect_date_col(df):
    for c in DATE_CANDIDATES:
        if c in df.columns:
            return c
    return None


def parse_date_series(s):
    s = s.astype(str).str.strip()
    for fr, en in _FR_MONTH.items():
        s = s.str.replace(fr, en, regex=False)
    return pd.to_datetime(s, errors="coerce", dayfirst=True)


def dedup_by_key_keep_latest(df, keys=None):
    df = df.copy()

    if keys is None:
        keys = [k for k in ["PATNO", "EVENT_ID"] if k in df.columns]

    if len(keys) == 0:
        return df.drop_duplicates(keep="last")

    dcol = detect_date_col(df)

    if dcol is not None:
        df["_DATE_TMP_"] = parse_date_series(df[dcol])
        df = df.sort_values(keys + ["_DATE_TMP_"])
        df = df.drop_duplicates(keys, keep="last")
        df = df.drop(columns=["_DATE_TMP_"])
    else:
        df = df.drop_duplicates(keys, keep="last")

    return df


def prefix_features(df, prefix, protected=("PATNO", "EVENT_ID")):
    df = df.copy()
    new_cols = {}
    for c in df.columns:
        if c not in protected:
            new_cols[c] = f"{prefix}{c}"
    return df.rename(columns=new_cols)


def event_id_to_month(x):
    x = str(x).strip().upper()
    if x == "BL":
        return 0
    m = re.fullmatch(r"V(\d+)", x)
    if m:
        v = int(m.group(1))
        return (v - 2) * 6
    return np.nan


def drop_admin_cols_safe(df):
    cols = []
    for c in df.columns:
        cu = c.upper()
        if cu in KEYS:
            continue
        if any(w in cu for w in ADMIN_WORDS):
            cols.append(c)
    return df.drop(columns=cols, errors="ignore")


def drop_leaky_cols_safe(df):
    cols = []
    for c in df.columns:
        cu = c.upper()
        if cu in KEYS:
            continue
        if any(w in cu for w in LEAKY_CONTAINS):
            cols.append(c)
    return df.drop(columns=cols, errors="ignore")


def filter_to_pd(participant_status):
    df = participant_status.copy()
    candidate_cols = [c for c in df.columns if "COHORT" in c or "STATUS" in c or "DEF" in c]

    if len(candidate_cols) == 0:
        txt = df.astype(str).agg(" ".join, axis=1)
    else:
        txt = pd.Series("", index=df.index)
        for c in candidate_cols:
            txt = txt.astype(str) + " " + df[c].astype(str)

    mask = txt.str.contains("PARKINSON", case=False, na=False)
    keep_pat = df.loc[mask, "PATNO"].dropna().astype(int).unique()
    return set(keep_pat)




# ==========================================================
# 5) LOAD CLINICAL FILES
# ==========================================================
demo_path   = find_file("Demographics.csv")
ess_path    = find_file("Epworth_Sleepiness_Scale.csv")
up1_path    = find_file("MDS-UPDRS_Part_I.csv")
up4_path    = find_file("MDS-UPDRS_Part_IV.csv")
rem_path    = find_file("REM_Sleep_Behavior.csv")
scopa_path  = find_file("SCOPA-AUT.csv")
gds_path    = find_file("Geriatric_Depression_Scale.csv")
status_path = find_file("Participant_Status.csv")
up23_path   = find_file("UPDRS_PartII_PartIII.xlsx")
mri_path    = find_mri_feature_file()

demo = prefix_features(
    dedup_by_key_keep_latest(
        coerce_keys(norm_cols(read_csv_robust(demo_path))),
        keys=["PATNO", "EVENT_ID"]
    ),
    "DEMO__"
)

ess = prefix_features(
    dedup_by_key_keep_latest(
        coerce_keys(norm_cols(read_csv_robust(ess_path))),
        keys=["PATNO", "EVENT_ID"]
    ),
    "ESS__"
)

up1 = prefix_features(
    dedup_by_key_keep_latest(
        coerce_keys(norm_cols(read_csv_robust(up1_path))),
        keys=["PATNO", "EVENT_ID"]
    ),
    "UP1__"
)

up4 = prefix_features(
    dedup_by_key_keep_latest(
        coerce_keys(norm_cols(read_csv_robust(up4_path))),
        keys=["PATNO", "EVENT_ID"]
    ),
    "UP4__"
)

rem = prefix_features(
    dedup_by_key_keep_latest(
        coerce_keys(norm_cols(read_csv_robust(rem_path))),
        keys=["PATNO", "EVENT_ID"]
    ),
    "REM__"
)

scopa = prefix_features(
    dedup_by_key_keep_latest(
        coerce_keys(norm_cols(read_csv_robust(scopa_path))),
        keys=["PATNO", "EVENT_ID"]
    ),
    "SCOPA__"
)

gds = prefix_features(
    dedup_by_key_keep_latest(
        coerce_keys(norm_cols(read_csv_robust(gds_path))),
        keys=["PATNO", "EVENT_ID"]
    ),
    "GDS__"
)

status = dedup_by_key_keep_latest(
    coerce_keys(norm_cols(read_csv_robust(status_path))),
    keys=["PATNO"]
)

up23 = pd.read_excel(up23_path)
up23 = prefix_features(
    dedup_by_key_keep_latest(
        coerce_keys(norm_cols(up23)),
        keys=["PATNO", "EVENT_ID"]
    ),
    "UP23__"
)

demo  = drop_leaky_cols_safe(drop_admin_cols_safe(demo))
ess   = drop_leaky_cols_safe(drop_admin_cols_safe(ess))
up1   = drop_leaky_cols_safe(drop_admin_cols_safe(up1))
up4   = drop_leaky_cols_safe(drop_admin_cols_safe(up4))
rem   = drop_leaky_cols_safe(drop_admin_cols_safe(rem))
scopa = drop_leaky_cols_safe(drop_admin_cols_safe(scopa))
gds   = drop_leaky_cols_safe(drop_admin_cols_safe(gds))
up23  = drop_leaky_cols_safe(drop_admin_cols_safe(up23))

dfs = [demo, ess, up1, up4, rem, scopa, gds, up23]
df_seq = reduce(lambda left, right: pd.merge(left, right, on=KEYS, how="outer"), dfs)

status = coerce_keys(norm_cols(status))
pd_patients = filter_to_pd(status)
df_seq = df_seq[df_seq["PATNO"].isin(pd_patients)].copy()

df_seq["MONTH"] = df_seq["EVENT_ID"].apply(event_id_to_month)
df_seq = df_seq[df_seq["MONTH"].notna()].copy()
df_seq["MONTH"] = df_seq["MONTH"].astype(int)

df_seq = df_seq.sort_values(["PATNO", "MONTH"])
df_seq = df_seq.drop_duplicates(subset=["PATNO", "MONTH"], keep="last").copy()

print("Merged df_seq shape after duplicate PATNO-MONTH removal:", df_seq.shape)




# ==========================================================
# 6) LABELS
# ==========================================================
if "GDS__DEPRESSION" not in df_seq.columns:
    gds_score_candidates = [c for c in df_seq.columns if "GDS" in c and ("TOTAL" in c or "SCORE" in c)]
    if len(gds_score_candidates) == 0:
        raise ValueError("GDS__DEPRESSION not found and no GDS score column detected.")
    gds_score_col = gds_score_candidates[0]
    score = pd.to_numeric(df_seq[gds_score_col], errors="coerce")
    df_seq["GDS__DEPRESSION"] = (score >= 5).astype("Float64")

if "GDS__LEVELS" not in df_seq.columns:
    gds_score_candidates = [c for c in df_seq.columns if "GDS" in c and ("TOTAL" in c or "SCORE" in c)]
    if len(gds_score_candidates) == 0:
        raise ValueError("GDS__LEVELS not found and no GDS score column detected.")
    gds_score_col = gds_score_candidates[0]
    score = pd.to_numeric(df_seq[gds_score_col], errors="coerce")

    levels = pd.Series(np.nan, index=df_seq.index)
    levels[(score >= 0) & (score <= 4)] = 0
    levels[(score >= 5) & (score <= 8)] = 1
    levels[(score >= 9) & (score <= 11)] = 2
    levels[(score >= 12)] = 3
    df_seq["GDS__LEVELS"] = levels.astype("Float64")

if "SUBTYPE_BIN" not in df_seq.columns:
    ratio_candidates = [c for c in df_seq.columns if "RATIO" in c.upper()]
    td_candidates = [c for c in df_seq.columns if "SCORE TD" in c.upper()]
    pigd_candidates = [c for c in df_seq.columns if "SCORE PIGD" in c.upper()]

    subtype = pd.Series(np.nan, index=df_seq.index)

    if len(ratio_candidates) > 0:
        ratio = pd.to_numeric(df_seq[ratio_candidates[0]], errors="coerce")
        subtype[ratio >= 1.15] = 1
        subtype[ratio <= 0.90] = 2
    elif len(td_candidates) > 0 and len(pigd_candidates) > 0:
        td = pd.to_numeric(df_seq[td_candidates[0]], errors="coerce")
        pigd = pd.to_numeric(df_seq[pigd_candidates[0]], errors="coerce")
        ratio = td / pigd.replace(0, np.nan)
        subtype[ratio >= 1.15] = 1
        subtype[ratio <= 0.90] = 2
    else:
        raise ValueError("Impossible de construire SUBTYPE_BIN")

    df_seq["SUBTYPE_BIN"] = subtype.astype("Float64")

print("Labels ready.")




# ==========================================================
# 7) KEEP NUMERIC CLINICAL FEATURES
# ==========================================================
protected_cols = [
    "PATNO", "EVENT_ID", "MONTH",
    "GDS__DEPRESSION", "GDS__LEVELS", "SUBTYPE_BIN"
]

for c in df_seq.columns:
    if c not in ["PATNO", "EVENT_ID"]:
        df_seq[c] = pd.to_numeric(df_seq[c], errors="ignore")

numeric_cols = []
for c in df_seq.columns:
    if c in protected_cols:
        continue
    if pd.api.types.is_numeric_dtype(df_seq[c]):
        numeric_cols.append(c)

feature_cols_model = []
for c in numeric_cols:
    miss = df_seq[c].isna().mean()
    protected_prefix = c.startswith(("ESS__", "REM__", "SCOPA__", "GDS__", "UP1__", "UP4__", "UP23__", "DEMO__"))
    if miss <= 0.90 or protected_prefix:
        feature_cols_model.append(c)

print("Clinical feature columns:", len(feature_cols_model))




# ==========================================================
# 8) LOAD PRE-EXTRACTED MRI FEATURES
# ==========================================================
def load_preextracted_mri_features(mri_file):
    mri_df = pd.read_csv(mri_file)
    mri_df.columns = [str(c).strip() for c in mri_df.columns]

    if "PATNO" not in mri_df.columns:
        raise ValueError("MRI file must contain PATNO column")

    mri_df["PATNO"] = pd.to_numeric(mri_df["PATNO"], errors="coerce").astype("Int64")

    feat_cols = [c for c in mri_df.columns if c.lower().startswith("mri_feat")]
    if len(feat_cols) != MRI_FEAT_DIM:
        raise ValueError(f"Expected {MRI_FEAT_DIM} MRI features, but found {len(feat_cols)}.")

    if "mri_available" not in mri_df.columns:
        mri_df["mri_available"] = 1

    mri_df = mri_df.dropna(subset=["PATNO"]).copy()
    mri_df["PATNO"] = mri_df["PATNO"].astype(int)

    feat_dict = {}
    for _, row in mri_df.iterrows():
        patno = int(row["PATNO"])
        available = float(row["mri_available"])

        if available > 0.5:
            feat = row[feat_cols].to_numpy(dtype=np.float32)
            feat_dict[patno] = {
                "feat": feat,
                "available": 1.0
            }
        else:
            feat_dict[patno] = {
                "feat": np.full(MRI_FEAT_DIM, MRI_MISSING_FILL, dtype=np.float32),
                "available": 0.0
            }

    print("MRI patients loaded:", len(feat_dict))
    print("MRI feature dim:", len(feat_cols))
    return feat_dict, feat_cols


mri_feat_dict, MRI_FEATURE_COLUMNS = load_preextracted_mri_features(mri_path)




# ==========================================================
# 9) MULTITASK SEQUENCES WITH MRI
# ==========================================================
def make_multitask_sequences(df_seq, feature_cols, months, last_month):
    Xc_list, Xmri_list, Xflag_list, pat_list = [], [], [], []
    y_subtype_list, m_subtype_list = [], []
    y_depbin_list, m_depbin_list = [], []
    y_levels_list, m_levels_list = [], []
    month_idx_list = []

    for patno, g in df_seq.groupby("PATNO"):
        g = g.copy()
        g = g.sort_values("MONTH")
        g = g.drop_duplicates(subset=["MONTH"], keep="last")
        g = g.set_index("MONTH")

        if last_month not in g.index:
            continue

        gg = g.reindex(months)
        Xc = gg[feature_cols].to_numpy(dtype=np.float32)
        month_idx = np.array([MONTH_TO_INDEX[m] for m in months], dtype=np.int64)

        row_last = g.loc[last_month]
        if isinstance(row_last, pd.DataFrame):
            row_last = row_last.iloc[-1]

        y_subtype = row_last["SUBTYPE_BIN"] if "SUBTYPE_BIN" in row_last.index else pd.NA
        y_depbin  = row_last["GDS__DEPRESSION"] if "GDS__DEPRESSION" in row_last.index else pd.NA
        y_levels  = row_last["GDS__LEVELS"] if "GDS__LEVELS" in row_last.index else pd.NA

        has_any = (
            (not pd.isna(y_subtype)) or
            (not pd.isna(y_depbin)) or
            (not pd.isna(y_levels))
        )
        if not has_any:
            continue

        if int(patno) in mri_feat_dict:
            xmri = mri_feat_dict[int(patno)]["feat"].astype(np.float32)
            xflag = np.array([mri_feat_dict[int(patno)]["available"]], dtype=np.float32)
        else:
            xmri = np.full(MRI_FEAT_DIM, MRI_MISSING_FILL, dtype=np.float32)
            xflag = np.array([0.0], dtype=np.float32)

        Xc_list.append(Xc)
        Xmri_list.append(xmri)
        Xflag_list.append(xflag)
        pat_list.append(int(patno))
        month_idx_list.append(month_idx)

        if pd.isna(y_subtype):
            y_subtype_list.append(-1)
            m_subtype_list.append(0.0)
        else:
            val = int(y_subtype)
            if val in [1, 2]:
                val = val - 1
            y_subtype_list.append(val)
            m_subtype_list.append(1.0)

        if pd.isna(y_depbin):
            y_depbin_list.append(-1)
            m_depbin_list.append(0.0)
        else:
            y_depbin_list.append(int(y_depbin))
            m_depbin_list.append(1.0)

        if pd.isna(y_levels):
            y_levels_list.append(-1)
            m_levels_list.append(0.0)
        else:
            y_levels_list.append(int(y_levels))
            m_levels_list.append(1.0)

    if len(Xc_list) == 0:
        raise ValueError("Aucune séquence construite.")

    return {
        "X": np.stack(Xc_list).astype(np.float32),
        "Xmri": np.stack(Xmri_list).astype(np.float32),
        "Xflag": np.stack(Xflag_list).astype(np.float32),
        "month_idx": np.stack(month_idx_list).astype(np.int64),
        "pat": np.array(pat_list, dtype=np.int64),
        "y_subtype": np.array(y_subtype_list, dtype=np.int64),
        "m_subtype": np.array(m_subtype_list, dtype=np.float32),
        "y_depbin": np.array(y_depbin_list, dtype=np.int64),
        "m_depbin": np.array(m_depbin_list, dtype=np.float32),
        "y_levels": np.array(y_levels_list, dtype=np.int64),
        "m_levels": np.array(m_levels_list, dtype=np.float32),
    }




# ==========================================================
# 10) PREPROCESS CLINICAL
# ==========================================================
def fit_preprocess_train_only_with_mask(X_train):
    X_train = X_train.astype(np.float32)
    N, T, F = X_train.shape

    mask = np.isfinite(X_train).astype(np.float32)

    flat = X_train.reshape(-1, F)
    med = np.nanmedian(flat, axis=0)
    med = np.where(np.isfinite(med), med, 0.0)

    flat_imp = flat.copy()
    inds = np.where(~np.isfinite(flat_imp))
    flat_imp[inds] = np.take(med, inds[1])

    scaler = MinMaxScaler()
    flat_scaled = scaler.fit_transform(flat_imp)
    X_scaled = flat_scaled.reshape(N, T, F).astype(np.float32)

    X_out = np.concatenate([X_scaled, mask], axis=-1).astype(np.float32)
    return X_out, med.astype(np.float32), scaler


def transform_with_train_params_with_mask(X, med, scaler):
    X = X.astype(np.float32)
    N, T, F = X.shape

    mask = np.isfinite(X).astype(np.float32)

    flat = X.reshape(-1, F)
    flat_imp = flat.copy()
    inds = np.where(~np.isfinite(flat_imp))
    flat_imp[inds] = np.take(med, inds[1])

    flat_scaled = scaler.transform(flat_imp)
    X_scaled = flat_scaled.reshape(N, T, F).astype(np.float32)

    X_out = np.concatenate([X_scaled, mask], axis=-1).astype(np.float32)
    return X_out




# ==========================================================
# 10-bis) PREPROCESS MRI
# ==========================================================
def fit_preprocess_mri_train_only(Xmri_train, Xflag_train, missing_fill=MRI_MISSING_FILL):
    Xmri_train = Xmri_train.astype(np.float32)
    available_mask = (Xflag_train.reshape(-1) > 0.5)

    scaler = StandardScaler()

    if available_mask.sum() > 1:
        scaler.fit(Xmri_train[available_mask])
        Xmri_scaled = Xmri_train.copy()
        Xmri_scaled[available_mask] = scaler.transform(Xmri_scaled[available_mask])
    else:
        scaler.fit(np.zeros((2, Xmri_train.shape[1]), dtype=np.float32))
        Xmri_scaled = Xmri_train.copy()

    Xmri_scaled[~available_mask] = missing_fill
    return Xmri_scaled.astype(np.float32), scaler


def transform_mri_with_train_params(Xmri, Xflag, scaler, missing_fill=MRI_MISSING_FILL):
    Xmri = Xmri.astype(np.float32)
    available_mask = (Xflag.reshape(-1) > 0.5)
    Xmri_scaled = Xmri.copy()

    if available_mask.sum() > 0:
        Xmri_scaled[available_mask] = scaler.transform(Xmri_scaled[available_mask])

    Xmri_scaled[~available_mask] = missing_fill
    return Xmri_scaled.astype(np.float32)




# ==========================================================
# 11) LOSSES
# ==========================================================
def inverse_freq_weights(y, n_classes, power=1.0, max_w=8.0):
    y = np.asarray(y, dtype=np.int64)
    counts = np.bincount(y, minlength=n_classes).astype(np.float32)
    counts[counts == 0] = 1.0
    w = (counts.sum() / counts) ** power
    w = w / w.mean()
    w = np.clip(w, 1.0, max_w)
    return torch.tensor(w, dtype=torch.float32, device=device)


class FocalLoss(nn.Module):
    def __init__(self, gamma=2.0, weight=None, reduction="mean"):
        super().__init__()
        self.gamma = gamma
        self.weight = weight
        self.reduction = reduction

    def forward(self, logits, targets):
        ce = F.cross_entropy(logits, targets, weight=self.weight, reduction="none")
        pt = torch.exp(-ce)
        loss = ((1 - pt) ** self.gamma) * ce

        if self.reduction == "mean":
            return loss.mean()
        if self.reduction == "sum":
            return loss.sum()
        return loss


def masked_focal_loss(logits, y, mask, focal_loss_fn):
    valid = mask > 0.5
    if valid.sum() == 0:
        return logits.sum() * 0.0
    return focal_loss_fn(logits[valid], y[valid])




# ==========================================================
# 12) DATA SLICING & SAMPLER
# ==========================================================
def slice_multitask_data(data, idx):
    return {
        "X": data["X"][idx],
        "Xmri": data["Xmri"][idx],
        "Xflag": data["Xflag"][idx],
        "month_idx": data["month_idx"][idx],
        "pat": data["pat"][idx],
        "y_subtype": data["y_subtype"][idx],
        "m_subtype": data["m_subtype"][idx],
        "y_depbin": data["y_depbin"][idx],
        "m_depbin": data["m_depbin"][idx],
        "y_levels": data["y_levels"][idx],
        "m_levels": data["m_levels"][idx],
    }


def make_weights_and_sampler(train_data, power=1.2):
    valid_sub = train_data["m_subtype"] > 0.5
    valid_dep = train_data["m_depbin"] > 0.5
    valid_lvl = train_data["m_levels"] > 0.5

    w_sub = inverse_freq_weights(train_data["y_subtype"][valid_sub], 2, power=power)
    w_dep = inverse_freq_weights(train_data["y_depbin"][valid_dep], 2, power=power)
    w_lvl = inverse_freq_weights(train_data["y_levels"][valid_lvl], 4, power=power)

    sample_w = np.ones(len(train_data["pat"]), dtype=np.float32)

    if valid_sub.sum() > 0:
        cnt = np.bincount(train_data["y_subtype"][valid_sub], minlength=2).astype(np.float32)
        cnt[cnt == 0] = 1.0
        ws = 1.0 / (cnt ** 0.9)
        sample_w += np.where(valid_sub, ws[np.clip(train_data["y_subtype"], 0, 1)], 0.0)

    if valid_dep.sum() > 0:
        cnt = np.bincount(train_data["y_depbin"][valid_dep], minlength=2).astype(np.float32)
        cnt[cnt == 0] = 1.0
        wd = 1.0 / (cnt ** 0.9)
        sample_w += np.where(valid_dep, wd[np.clip(train_data["y_depbin"], 0, 1)], 0.0)

    if valid_lvl.sum() > 0:
        cnt = np.bincount(train_data["y_levels"][valid_lvl], minlength=4).astype(np.float32)
        cnt[cnt == 0] = 1.0
        wl = 1.0 / (cnt ** 0.95)
        sample_w += np.where(valid_lvl, wl[np.clip(train_data["y_levels"], 0, 3)], 0.0)

    sample_w = torch.tensor(sample_w, dtype=torch.float32)
    sampler = WeightedRandomSampler(sample_w, num_samples=len(sample_w), replacement=True)

    return w_sub, w_dep, w_lvl, sampler


# ==========================================================
# 13) MODEL
# PARALLEL FUSION VERSION (variante à 3 branches)
# ----------------------------------------------------------
# Clinical sequence -> shared embedding z
#   Branch A (Transformer encoder)  : z -> z_tr  -> tr_ctx  (last token)
#   Branch B (BiGRU)                : z -> z_gru -> g_last  (last hidden, both directions)
#   -> Transformer and BiGRU are run IN PARALLEL on the SAME input z
#      (the BiGRU no longer consumes the Transformer output: NOT sequential)
#   Branch C (MRI projector)        : x_mri -> mri_ctx
#
# Final fusion (3-branch, late fusion at the end):
#   [tr_ctx , g_last , mri_ctx , mri_flag] -> fusion MLP -> shared_dim -> 3 task heads
# ==========================================================
class TaskHead(nn.Module):
    def __init__(self, in_dim, hidden_dim, out_dim, dropout=0.3):
        super().__init__()
        self.net = nn.Sequential(
            nn.Linear(in_dim, hidden_dim),
            nn.LayerNorm(hidden_dim),
            nn.GELU(),
            nn.Dropout(dropout),
            nn.Linear(hidden_dim, out_dim)
        )

    def forward(self, x):
        return self.net(x)


class MRIProjector(nn.Module):
    def __init__(self, in_dim=256, hidden_dim=128, out_dim=64, dropout=0.25):
        super().__init__()
        self.net = nn.Sequential(
            nn.Linear(in_dim, hidden_dim),
            nn.LayerNorm(hidden_dim),
            nn.GELU(),
            nn.Dropout(dropout),
            nn.Linear(hidden_dim, out_dim),
            nn.LayerNorm(out_dim),
            nn.GELU(),
            nn.Dropout(dropout)
        )

    def forward(self, x):
        return self.net(x)


class ParallelTransformerGRUFusion(nn.Module):
    """
    Multitask hybrid model with Transformer and BiGRU running IN PARALLEL
    (both consume the same embedded clinical sequence `z`, independently),
    fused at the very end with the MRI branch into a 3-vector concatenation:
        [tr_ctx, g_last, mri_ctx] (+ mri_flag)
    """

    def __init__(
        self,
        clinical_input_dim,
        mri_input_dim=256,
        mri_proj_dim=64,
        d_model=128,
        nhead=4,
        num_layers=2,
        dim_feedforward=256,
        gru_hidden=64,
        gru_layers=1,
        shared_dim=128,
        dropout=0.30,
        input_noise_std=0.01,
    ):
        super().__init__()

        self.input_noise_std = input_noise_std

        # ======================================================
        # Shared clinical embedding (input to BOTH parallel branches)
        # ======================================================
        self.time_embed = nn.Embedding(6, d_model)
        self.input_proj = nn.Linear(clinical_input_dim, d_model)
        self.input_norm = nn.LayerNorm(d_model)

        # ======================================================
        # Branch A: Transformer encoder (parallel branch)
        # ======================================================
        encoder_layer = nn.TransformerEncoderLayer(
            d_model=d_model,
            nhead=nhead,
            dim_feedforward=dim_feedforward,
            dropout=dropout,
            batch_first=True,
            activation="gelu"
        )
        self.transformer = nn.TransformerEncoder(encoder_layer, num_layers=num_layers)
        tr_ctx_dim = d_model

        # ======================================================
        # Branch B: BiGRU (parallel branch - consumes `z`, NOT z_tr)
        # ======================================================
        self.gru = nn.GRU(
            input_size=d_model,
            hidden_size=gru_hidden,
            num_layers=gru_layers,
            batch_first=True,
            bidirectional=True,
            dropout=dropout if gru_layers > 1 else 0.0
        )
        g_last_dim = 2 * gru_hidden

        # ======================================================
        # Branch C: MRI projector (parallel branch)
        # ======================================================
        self.mri_projector = MRIProjector(
            in_dim=mri_input_dim,
            hidden_dim=max(96, mri_proj_dim * 2),
            out_dim=mri_proj_dim,
            dropout=dropout
        )
        mri_ctx_dim = mri_proj_dim

        # ======================================================
        # Final 3-branch fusion: tr_ctx + g_last + mri_ctx (+ flag)
        # ======================================================
        fusion_dim = tr_ctx_dim + g_last_dim + mri_ctx_dim + 1

        self.fusion = nn.Sequential(
            nn.Linear(fusion_dim, shared_dim),
            nn.LayerNorm(shared_dim),
            nn.GELU(),
            nn.Dropout(dropout),
            nn.Linear(shared_dim, shared_dim),
            nn.LayerNorm(shared_dim),
            nn.GELU(),
            nn.Dropout(dropout),
        )

        self.head_subtype = TaskHead(shared_dim, 64, 2, dropout)
        self.head_depbin  = TaskHead(shared_dim, 64, 2, dropout)
        self.head_levels  = TaskHead(shared_dim, 96, 4, dropout)

        # Multitask homoscedastic uncertainty weights (same as before)
        self.log_vars = nn.Parameter(torch.zeros(3))

    def forward(self, x_clin, x_mri, x_flag, month_idx):
        # ======================================================
        # 0) Shared embedding
        # ======================================================
        x = x_clin
        if self.training and self.input_noise_std > 0:
            noise = torch.randn_like(x) * self.input_noise_std
            x = x + noise

        z = self.input_proj(x)
        z = self.input_norm(z)
        z = z + self.time_embed(month_idx)

        # ======================================================
        # 1) PARALLEL branches (both read `z`, independently)
        # ======================================================
        z_tr = self.transformer(z)          # Branch A: Transformer
        tr_ctx = z_tr[:, -1, :]             # context vector from transformer

        z_gru, _ = self.gru(z)              # Branch B: BiGRU (reads z, NOT z_tr)
        g_last = z_gru[:, -1, :]            # last hidden state (both directions)

        mri_ctx = self.mri_projector(x_mri)  # Branch C: MRI

        # ======================================================
        # 2) Fusion at the end: 3 vectors [tr_ctx, g_last, mri_ctx] + flag
        # ======================================================
        fused_features = torch.cat([tr_ctx, g_last, mri_ctx, x_flag], dim=1)
        z_shared = self.fusion(fused_features)

        # ======================================================
        # 3) Classification heads
        # ======================================================
        out_subtype = self.head_subtype(z_shared)
        out_depbin  = self.head_depbin(z_shared)
        out_levels  = self.head_levels(z_shared)

        return out_subtype, out_depbin, out_levels


def build_model(params, clinical_input_dim):
    return ParallelTransformerGRUFusion(
        clinical_input_dim=clinical_input_dim,
        mri_input_dim=MRI_FEAT_DIM,
        mri_proj_dim=params["mri_proj_dim"],
        d_model=params["d_model"],
        nhead=params["nhead"],
        num_layers=params["num_layers"],
        dim_feedforward=params["dim_feedforward"],
        gru_hidden=params["gru_hidden"],
        gru_layers=params["gru_layers"],
        shared_dim=params["shared_dim"],
        dropout=params["dropout"],
        input_noise_std=params["input_noise_std"],
    ).to(device)




# ==========================================================
# 14) DATALOADER
# ==========================================================
def build_loader(Xc, Xmri, Xflag, Midx, ys, ms, yd, md, yl, ml, batch_size=32, shuffle=False, sampler=None):
    ds = TensorDataset(
        torch.tensor(Xc, dtype=torch.float32),
        torch.tensor(Xmri, dtype=torch.float32),
        torch.tensor(Xflag, dtype=torch.float32),
        torch.tensor(Midx, dtype=torch.long),
        torch.tensor(ys, dtype=torch.long),
        torch.tensor(ms, dtype=torch.float32),
        torch.tensor(yd, dtype=torch.long),
        torch.tensor(md, dtype=torch.float32),
        torch.tensor(yl, dtype=torch.long),
        torch.tensor(ml, dtype=torch.float32),
    )
    return DataLoader(
        ds,
        batch_size=batch_size,
        shuffle=shuffle if sampler is None else False,
        sampler=sampler,
        num_workers=NUM_WORKERS,
        pin_memory=PIN_MEMORY
    )




# ==========================================================
# 15) REPORTING
# ==========================================================
def masked_metrics_full(y_true, y_pred, y_mask, n_classes, class_names):
    y_true = np.asarray(y_true)
    y_pred = np.asarray(y_pred)
    y_mask = np.asarray(y_mask).astype(bool)

    if y_mask.sum() == 0:
        return {
            "accuracy": np.nan,
            "precision_macro": np.nan,
            "recall_macro": np.nan,
            "f1_macro": np.nan,
            "balanced_accuracy": np.nan,
            "confusion_matrix": np.full((n_classes, n_classes), np.nan).tolist(),
            "classification_report": "No valid data.",
            "per_class": {c: {"precision": np.nan, "recall": np.nan, "f1": np.nan} for c in class_names},
        }

    yt = y_true[y_mask]
    yp = y_pred[y_mask]
    labels = list(range(n_classes))

    report_dict = classification_report(
        yt, yp, labels=labels, target_names=class_names,
        digits=4, zero_division=0, output_dict=True
    )
    per_class = {
        cname: {
            "precision": float(report_dict[cname]["precision"]),
            "recall": float(report_dict[cname]["recall"]),
            "f1": float(report_dict[cname]["f1-score"]),
        }
        for cname in class_names
    }

    return {
        "accuracy": float(accuracy_score(yt, yp)),
        "precision_macro": float(precision_score(yt, yp, labels=labels, average="macro", zero_division=0)),
        "recall_macro": float(recall_score(yt, yp, labels=labels, average="macro", zero_division=0)),
        "f1_macro": float(f1_score(yt, yp, labels=labels, average="macro", zero_division=0)),
        "balanced_accuracy": float(balanced_accuracy_score(yt, yp)),
        "confusion_matrix": confusion_matrix(yt, yp, labels=labels).tolist(),
        "classification_report": classification_report(yt, yp, labels=labels, target_names=class_names, digits=4, zero_division=0),
        "per_class": per_class,
    }


def plot_confusion_matrix_on_ax(cm, class_names, ax, title="Confusion Matrix"):
    cm = np.asarray(cm)
    disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=class_names)
    disp.plot(ax=ax, cmap="Blues", colorbar=False, values_format=".0f")
    ax.set_title(title, fontsize=11)


def plot_confusion_matrix_from_arrays(y_true, y_pred, y_mask, class_names, out_path, title="Confusion Matrix"):
    y_true = np.asarray(y_true)
    y_pred = np.asarray(y_pred)
    y_mask = np.asarray(y_mask).astype(bool)

    if y_mask.sum() == 0:
        print(f"[SKIP] No valid data for {title}")
        return []

    yt = y_true[y_mask]
    yp = y_pred[y_mask]

    cm = confusion_matrix(yt, yp, labels=list(range(len(class_names))))

    fig, ax = plt.subplots(figsize=(5.5, 5.5))
    plot_confusion_matrix_on_ax(cm, class_names, ax, title=title)
    plt.tight_layout()
    plt.savefig(out_path, dpi=150)
    plt.close()

    return cm.tolist()




# ==========================================================
# 16) EVALUATION
# ==========================================================
def evaluate_model(model, loader):
    model.eval()

    all_sub_t, all_sub_p, all_sub_m = [], [], []
    all_dep_t, all_dep_p, all_dep_m = [], [], []
    all_lvl_t, all_lvl_p, all_lvl_m = [], [], []

    with torch.no_grad():
        for batch in loader:
            x_clin, x_mri, x_flag, midx, ys, ms, yd, md, yl, ml = [b.to(device) for b in batch]
            out_s, out_d, out_l = model(x_clin, x_mri, x_flag, midx)

            pred_s = out_s.argmax(dim=1)
            pred_d = out_d.argmax(dim=1)
            pred_l = out_l.argmax(dim=1)

            all_sub_t.extend(ys.cpu().numpy().tolist())
            all_sub_p.extend(pred_s.cpu().numpy().tolist())
            all_sub_m.extend(ms.cpu().numpy().tolist())

            all_dep_t.extend(yd.cpu().numpy().tolist())
            all_dep_p.extend(pred_d.cpu().numpy().tolist())
            all_dep_m.extend(md.cpu().numpy().tolist())

            all_lvl_t.extend(yl.cpu().numpy().tolist())
            all_lvl_p.extend(pred_l.cpu().numpy().tolist())
            all_lvl_m.extend(ml.cpu().numpy().tolist())

    res_sub = masked_metrics_full(all_sub_t, all_sub_p, all_sub_m, N_CLASSES["subtype"], CLASS_NAMES["subtype"])
    res_dep = masked_metrics_full(all_dep_t, all_dep_p, all_dep_m, N_CLASSES["depression"], CLASS_NAMES["depression"])
    res_lvl = masked_metrics_full(all_lvl_t, all_lvl_p, all_lvl_m, N_CLASSES["levels"], CLASS_NAMES["levels"])

    global_score = np.nanmean([
        res_sub["balanced_accuracy"], res_sub["f1_macro"],
        res_dep["balanced_accuracy"], res_dep["f1_macro"],
        res_lvl["balanced_accuracy"], res_lvl["f1_macro"]
    ])

    return {
        "subtype": res_sub,
        "depression": res_dep,
        "levels": res_lvl,
        "global_score": float(global_score)
    }




# ==========================================================
# 17) TRAIN ONE MODEL
# ==========================================================
def train_one_model(train_loader, val_loader, params, clinical_input_dim):
    model = build_model(params, clinical_input_dim=clinical_input_dim)

    optimizer = torch.optim.AdamW(
        model.parameters(),
        lr=params["lr"],
        weight_decay=params["weight_decay"]
    )

    scheduler = torch.optim.lr_scheduler.ReduceLROnPlateau(
        optimizer,
        mode="max",
        factor=0.5,
        patience=4,
        min_lr=1e-5
    )

    train_tensors = train_loader.dataset.tensors
    ys = train_tensors[4].cpu().numpy()
    ms = train_tensors[5].cpu().numpy()
    yd = train_tensors[6].cpu().numpy()
    md = train_tensors[7].cpu().numpy()
    yl = train_tensors[8].cpu().numpy()
    ml = train_tensors[9].cpu().numpy()

    w_sub = inverse_freq_weights(ys[ms > 0.5], 2, power=params["class_power"])
    w_dep = inverse_freq_weights(yd[md > 0.5], 2, power=params["class_power"])
    w_lvl = inverse_freq_weights(yl[ml > 0.5], 4, power=params["class_power"])

    focal_sub = FocalLoss(gamma=params["focal_gamma_sub"], weight=w_sub, reduction="mean")
    focal_dep = FocalLoss(gamma=params["focal_gamma_dep"], weight=w_dep, reduction="mean")
    focal_lvl = FocalLoss(gamma=params["focal_gamma_lvl"], weight=w_lvl, reduction="mean")

    best_state = None
    best_score = -np.inf
    no_improve = 0
    history = {"train_loss": [], "val_score": []}

    for epoch in range(FINAL_EPOCHS):
        model.train()
        batch_losses = []

        for batch in train_loader:
            x_clin, x_mri, x_flag, midx, ys_b, ms_b, yd_b, md_b, yl_b, ml_b = [b.to(device) for b in batch]
            optimizer.zero_grad(set_to_none=True)

            out_s, out_d, out_l = model(x_clin, x_mri, x_flag, midx)

            loss_s = masked_focal_loss(out_s, ys_b, ms_b, focal_sub)
            loss_d = masked_focal_loss(out_d, yd_b, md_b, focal_dep)
            loss_l = masked_focal_loss(out_l, yl_b, ml_b, focal_lvl)

            loss = (
                TASK_IMPORTANCE["subtype"]    * (torch.exp(-model.log_vars[0]) * loss_s + model.log_vars[0]) +
                TASK_IMPORTANCE["depression"] * (torch.exp(-model.log_vars[1]) * loss_d + model.log_vars[1]) +
                TASK_IMPORTANCE["levels"]     * (torch.exp(-model.log_vars[2]) * loss_l + model.log_vars[2])
            )

            loss.backward()
            nn.utils.clip_grad_norm_(model.parameters(), max_norm=1.0)
            optimizer.step()

            batch_losses.append(float(loss.item()))

        train_loss = float(np.mean(batch_losses)) if len(batch_losses) else np.nan
        val_eval = evaluate_model(model, val_loader)
        val_score = val_eval["global_score"]

        scheduler.step(val_score)

        history["train_loss"].append(train_loss)
        history["val_score"].append(val_score)

        print(f"Epoch {epoch+1:02d} | train_loss={train_loss:.4f} | val_score={val_score:.4f}")

        if val_score > best_score:
            best_score = val_score
            best_state = {k: v.detach().cpu().clone() for k, v in model.state_dict().items()}
            no_improve = 0
        else:
            no_improve += 1

        if no_improve >= PATIENCE:
            print("Early stopping.")
            break

    if best_state is not None:
        model.load_state_dict(best_state)

    return model, history, best_score




# ==========================================================
# 18) HYPERPARAMETER SEARCH (SINGLE SPLIT - NO FOLD)
# ==========================================================
def group_split_indices(pat_all, unique_pat, test_ratio, random_state):
    gss = GroupShuffleSplit(n_splits=1, test_size=test_ratio, random_state=random_state)
    tr_pat_idx, te_pat_idx = next(gss.split(unique_pat, groups=unique_pat))
    tr_set = set(unique_pat[tr_pat_idx])
    te_set = set(unique_pat[te_pat_idx])
    tr_idx = np.array([i for i, p in enumerate(pat_all) if p in tr_set], dtype=int)
    te_idx = np.array([i for i, p in enumerate(pat_all) if p in te_set], dtype=int)
    return tr_idx, te_idx


def run_hyperparameter_search(data, pat_all, unique_pat, n_trials=N_TRIALS):
    """
    Optuna search using a SINGLE train/val split (no k-fold): 'sans fold'.
    """
    tr_idx_full, _test_idx_unused = group_split_indices(pat_all, unique_pat, TEST_RATIO, random_state=SEED)
    train_full = slice_multitask_data(data, tr_idx_full)

    inner_unique_pat = np.unique(train_full["pat"])
    inner_split = GroupShuffleSplit(n_splits=1, test_size=INNER_VAL_RATIO, random_state=SEED)
    inner_tr_pat_idx, inner_va_pat_idx = next(inner_split.split(inner_unique_pat, groups=inner_unique_pat))

    inner_tr_set = set(inner_unique_pat[inner_tr_pat_idx])
    inner_va_set = set(inner_unique_pat[inner_va_pat_idx])

    inner_tr_idx = np.array([i for i, p in enumerate(train_full["pat"]) if p in inner_tr_set], dtype=int)
    inner_va_idx = np.array([i for i, p in enumerate(train_full["pat"]) if p in inner_va_set], dtype=int)

    train_data = slice_multitask_data(train_full, inner_tr_idx)
    val_data   = slice_multitask_data(train_full, inner_va_idx)

    Xtr_c, med_vec, scaler = fit_preprocess_train_only_with_mask(train_data["X"])
    Xva_c = transform_with_train_params_with_mask(val_data["X"], med_vec, scaler)

    Xmri_tr, scaler_mri = fit_preprocess_mri_train_only(train_data["Xmri"], train_data["Xflag"])
    Xmri_va = transform_mri_with_train_params(val_data["Xmri"], val_data["Xflag"], scaler_mri)

    clinical_input_dim = Xtr_c.shape[-1]

    def objective(trial):
        params = {
            "mri_proj_dim": trial.suggest_categorical("mri_proj_dim", [32, 48, 64]),
            "d_model": trial.suggest_categorical("d_model", [96, 128, 160]),
            "nhead": trial.suggest_categorical("nhead", [2, 4]),
            "num_layers": trial.suggest_int("num_layers", 1, 3),
            "dim_feedforward": trial.suggest_categorical("dim_feedforward", [192, 256, 384]),
            "gru_hidden": trial.suggest_categorical("gru_hidden", [48, 64, 96]),
            "gru_layers": trial.suggest_int("gru_layers", 1, 2),
            "shared_dim": trial.suggest_categorical("shared_dim", [128, 160, 192]),
            "dropout": trial.suggest_float("dropout", 0.20, 0.40),
            "input_noise_std": trial.suggest_float("input_noise_std", 0.0, 0.03),
            "lr": trial.suggest_float("lr", 3e-4, 1.5e-3, log=True),
            "weight_decay": trial.suggest_float("weight_decay", 1e-6, 1e-3, log=True),
            "class_power": trial.suggest_float("class_power", 1.0, 1.5),
            "focal_gamma_sub": trial.suggest_float("focal_gamma_sub", 1.0, 2.5),
            "focal_gamma_dep": trial.suggest_float("focal_gamma_dep", 1.0, 2.5),
            "focal_gamma_lvl": trial.suggest_float("focal_gamma_lvl", 1.5, 3.0),
        }

        w_sub, w_dep, w_lvl, sampler = make_weights_and_sampler(train_data, power=params["class_power"])

        train_loader = build_loader(
            Xtr_c, Xmri_tr, train_data["Xflag"], train_data["month_idx"],
            train_data["y_subtype"], train_data["m_subtype"],
            train_data["y_depbin"], train_data["m_depbin"],
            train_data["y_levels"], train_data["m_levels"],
            batch_size=32, sampler=sampler
        )

        val_loader = build_loader(
            Xva_c, Xmri_va, val_data["Xflag"], val_data["month_idx"],
            val_data["y_subtype"], val_data["m_subtype"],
            val_data["y_depbin"], val_data["m_depbin"],
            val_data["y_levels"], val_data["m_levels"],
            batch_size=64, shuffle=False
        )

        model = build_model(params, clinical_input_dim=clinical_input_dim)
        optimizer = torch.optim.AdamW(model.parameters(), lr=params["lr"], weight_decay=params["weight_decay"])
        scheduler = torch.optim.lr_scheduler.ReduceLROnPlateau(
            optimizer, mode="max", factor=0.5, patience=3, min_lr=1e-5
        )

        focal_sub = FocalLoss(gamma=params["focal_gamma_sub"], weight=w_sub, reduction="mean")
        focal_dep = FocalLoss(gamma=params["focal_gamma_dep"], weight=w_dep, reduction="mean")
        focal_lvl = FocalLoss(gamma=params["focal_gamma_lvl"], weight=w_lvl, reduction="mean")

        best_score = -np.inf
        no_improve = 0
        local_patience = 5

        for epoch in range(SEARCH_EPOCHS):
            model.train()

            for batch in train_loader:
                x_clin, x_mri, x_flag, midx, ys_b, ms_b, yd_b, md_b, yl_b, ml_b = [b.to(device) for b in batch]
                optimizer.zero_grad(set_to_none=True)

                out_s, out_d, out_l = model(x_clin, x_mri, x_flag, midx)

                loss_s = masked_focal_loss(out_s, ys_b, ms_b, focal_sub)
                loss_d = masked_focal_loss(out_d, yd_b, md_b, focal_dep)
                loss_l = masked_focal_loss(out_l, yl_b, ml_b, focal_lvl)

                loss = (
                    TASK_IMPORTANCE["subtype"]    * (torch.exp(-model.log_vars[0]) * loss_s + model.log_vars[0]) +
                    TASK_IMPORTANCE["depression"] * (torch.exp(-model.log_vars[1]) * loss_d + model.log_vars[1]) +
                    TASK_IMPORTANCE["levels"]     * (torch.exp(-model.log_vars[2]) * loss_l + model.log_vars[2])
                )

                loss.backward()
                nn.utils.clip_grad_norm_(model.parameters(), max_norm=1.0)
                optimizer.step()

            val_eval = evaluate_model(model, val_loader)
            val_score = val_eval["global_score"]
            scheduler.step(val_score)

            if val_score > best_score:
                best_score = val_score
                no_improve = 0
            else:
                no_improve += 1

            if no_improve >= local_patience:
                break

            trial.report(val_score, step=epoch)
            if trial.should_prune():
                raise optuna.TrialPruned()

        del model, train_loader, val_loader
        gc.collect()

        return float(best_score)

    study = optuna.create_study(
        direction="maximize",
        sampler=TPESampler(seed=SEED),
        pruner=MedianPruner(n_startup_trials=4, n_warmup_steps=2)
    )
    study.optimize(objective, n_trials=n_trials, show_progress_bar=False)

    return study




# ==========================================================
# 19) AGGREGATION UTILITIES OVER THE 5 SEEDS (NO FOLDS)
# ==========================================================
def aggregate_task_over_seeds(seed_evals, task_name, n_classes, class_names):
    """
    seed_evals: list of the `res_<task>` dict returned by masked_metrics_full,
    one per seed (length N_SEEDS).
    Returns mean/std for global metrics, per-class metrics, and the
    mean confusion matrix (rounded to nearest int).
    """
    accs   = [e["accuracy"] for e in seed_evals]
    precs  = [e["precision_macro"] for e in seed_evals]
    recs   = [e["recall_macro"] for e in seed_evals]
    f1s    = [e["f1_macro"] for e in seed_evals]
    baccs  = [e["balanced_accuracy"] for e in seed_evals]

    def mstd(vals):
        vals = np.array(vals, dtype=float)
        return float(np.nanmean(vals)), float(np.nanstd(vals))

    acc_m, acc_s   = mstd(accs)
    prec_m, prec_s = mstd(precs)
    rec_m, rec_s   = mstd(recs)
    f1_m, f1_s     = mstd(f1s)
    bacc_m, bacc_s = mstd(baccs)

    per_class_agg = {}
    for cname in class_names:
        p_vals = [e["per_class"][cname]["precision"] for e in seed_evals]
        r_vals = [e["per_class"][cname]["recall"] for e in seed_evals]
        f_vals = [e["per_class"][cname]["f1"] for e in seed_evals]
        p_m, p_s = mstd(p_vals)
        r_m, r_s = mstd(r_vals)
        f_m, f_s = mstd(f_vals)
        per_class_agg[cname] = {
            "precision_mean": p_m, "precision_std": p_s,
            "recall_mean": r_m, "recall_std": r_s,
            "f1_mean": f_m, "f1_std": f_s,
        }

    cms = np.array([np.array(e["confusion_matrix"], dtype=float) for e in seed_evals])  # (n_seeds, C, C)
    cm_mean = np.nanmean(cms, axis=0)
    cm_mean_rounded = np.round(cm_mean).astype(int)

    return {
        "accuracy_mean": acc_m, "accuracy_std": acc_s,
        "precision_macro_mean": prec_m, "precision_macro_std": prec_s,
        "recall_macro_mean": rec_m, "recall_macro_std": rec_s,
        "f1_macro_mean": f1_m, "f1_macro_std": f1_s,
        "balanced_accuracy_mean": bacc_m, "balanced_accuracy_std": bacc_s,
        "per_class": per_class_agg,
        "confusion_matrix_mean_rounded": cm_mean_rounded.tolist(),
    }


def aggregate_histories_over_seeds(histories):
    """
    histories: list of {"train_loss": [...], "val_score": [...]} dicts, one per seed.
    Pads with NaN to the max length across seeds, then computes nanmean/nanstd per epoch.
    """
    max_len = max(len(h["train_loss"]) for h in histories)

    train_mat = np.full((len(histories), max_len), np.nan, dtype=float)
    val_mat = np.full((len(histories), max_len), np.nan, dtype=float)

    for i, h in enumerate(histories):
        L = len(h["train_loss"])
        train_mat[i, :L] = h["train_loss"]
        val_mat[i, :L] = h["val_score"]

    mean_train = np.nanmean(train_mat, axis=0)
    std_train = np.nanstd(train_mat, axis=0)
    mean_val = np.nanmean(val_mat, axis=0)
    std_val = np.nanstd(val_mat, axis=0)

    return {
        "mean_train_loss": mean_train,
        "std_train_loss": std_train,
        "mean_val_score": mean_val,
        "std_val_score": std_val,
    }


def plot_confusion_figure_3horizons(task_name, class_names, cms_by_horizon, out_path):
    """
    One figure containing the 3 (mean, rounded) confusion matrices for T2/T4/T6.
    """
    h_names = list(HORIZONS.keys())
    fig, axes = plt.subplots(1, len(h_names), figsize=(5.2 * len(h_names), 5.0))
    if len(h_names) == 1:
        axes = [axes]

    for ax, h_name in zip(axes, h_names):
        cm = np.array(cms_by_horizon[h_name])
        plot_confusion_matrix_on_ax(cm, class_names, ax, title=f"{h_name} - {task_name.capitalize()}")

    fig.suptitle(f"Mean Confusion Matrices (rounded) over {N_SEEDS} seeds - {task_name.capitalize()}", fontsize=13)
    plt.tight_layout()
    plt.savefig(out_path, dpi=150)
    plt.show()
    plt.close()


def plot_loss_figure_3horizons(agg_histories_by_horizon, out_path):
    """
    One figure containing the 3 Train/Val loss curves (mean over seeds) for T2/T4/T6.
    """
    h_names = list(HORIZONS.keys())
    fig, axes = plt.subplots(1, len(h_names), figsize=(6.0 * len(h_names), 4.5))
    if len(h_names) == 1:
        axes = [axes]

    for ax, h_name in zip(axes, h_names):
        agg = agg_histories_by_horizon[h_name]
        epochs = np.arange(1, len(agg["mean_train_loss"]) + 1)

        ax.plot(epochs, agg["mean_train_loss"], label="Train Loss (mean)", color="tab:blue")
        ax.fill_between(
            epochs,
            agg["mean_train_loss"] - agg["std_train_loss"],
            agg["mean_train_loss"] + agg["std_train_loss"],
            color="tab:blue", alpha=0.2
        )

        ax.plot(epochs, agg["mean_val_score"], label="Val Score (mean)", color="tab:orange")
        ax.fill_between(
            epochs,
            agg["mean_val_score"] - agg["std_val_score"],
            agg["mean_val_score"] + agg["std_val_score"],
            color="tab:orange", alpha=0.2
        )

        ax.set_title(f"{h_name}")
        ax.set_xlabel("Epoch")
        ax.set_ylabel("Value")
        ax.legend()

    fig.suptitle(f"Mean Train Loss / Val Score over {N_SEEDS} seeds", fontsize=13)
    plt.tight_layout()
    plt.savefig(out_path, dpi=150)
    plt.show()
    plt.close()




# ==========================================================
# 20) RUN ONE HORIZON (5 SEEDS - REPEATED HOLD-OUT, NO FOLDS)
# ==========================================================
def run_one_horizon(h_name, months, last_month):
    print("\n" + "=" * 80)
    print(f"RUNNING HORIZON {h_name} | months={months} | last_month={last_month}")
    print("=" * 80)

    horizon_dir = OUT_DIR / h_name
    horizon_dir.mkdir(parents=True, exist_ok=True)

    data = make_multitask_sequences(df_seq, feature_cols_model, months=months, last_month=last_month)

    pat_all = data["pat"]
    unique_pat = np.unique(pat_all)

    print("Total sequences:", len(pat_all))
    print("Unique patients:", len(unique_pat))
    print("Clinical shape:", data["X"].shape, "| MRI shape:", data["Xmri"].shape)

    # ------------------------------------------------------------
    # HYPERPARAMETER SEARCH (single split, "sans fold")
    # ------------------------------------------------------------
    study = run_hyperparameter_search(data, pat_all, unique_pat, n_trials=N_TRIALS)
    best_params = study.best_trial.params
    print("\nBEST PARAMS:", best_params)
    print("BEST SEARCH SCORE:", study.best_value)

    with open(horizon_dir / "best_params.json", "w", encoding="utf-8") as f:
        json.dump(best_params, f, indent=2)

    # ------------------------------------------------------------
    # FINAL TRAINING / EVALUATION OVER 5 INDEPENDENT SEEDS
    # ------------------------------------------------------------
    seed_results = []      # list of evaluate_model() dicts, one per seed
    seed_histories = []    # list of training histories, one per seed
    all_predictions = []

    for seed_id, seed in enumerate(SEEDS, start=1):
        print("\n" + "-" * 70)
        print(f"{h_name} | SEED {seed_id}/{N_SEEDS} (random_state={seed})")
        print("-" * 70)

        set_seed(seed)

        seed_dir = horizon_dir / f"seed_{seed}"
        seed_dir.mkdir(parents=True, exist_ok=True)

        tr_idx_full, te_idx = group_split_indices(pat_all, unique_pat, TEST_RATIO, random_state=seed)
        train_full = slice_multitask_data(data, tr_idx_full)
        test_data = slice_multitask_data(data, te_idx)

        inner_unique_pat = np.unique(train_full["pat"])
        inner_split = GroupShuffleSplit(n_splits=1, test_size=INNER_VAL_RATIO, random_state=seed + 1000)
        inner_tr_pat_idx, inner_va_pat_idx = next(inner_split.split(inner_unique_pat, groups=inner_unique_pat))

        inner_tr_set = set(inner_unique_pat[inner_tr_pat_idx])
        inner_va_set = set(inner_unique_pat[inner_va_pat_idx])

        inner_tr_idx = np.array([i for i, p in enumerate(train_full["pat"]) if p in inner_tr_set], dtype=int)
        inner_va_idx = np.array([i for i, p in enumerate(train_full["pat"]) if p in inner_va_set], dtype=int)

        train_data = slice_multitask_data(train_full, inner_tr_idx)
        val_data   = slice_multitask_data(train_full, inner_va_idx)

        Xtr_c, med_vec, scaler = fit_preprocess_train_only_with_mask(train_data["X"])
        Xva_c = transform_with_train_params_with_mask(val_data["X"], med_vec, scaler)
        Xte_c = transform_with_train_params_with_mask(test_data["X"], med_vec, scaler)

        Xmri_tr, scaler_mri = fit_preprocess_mri_train_only(train_data["Xmri"], train_data["Xflag"])
        Xmri_va = transform_mri_with_train_params(val_data["Xmri"], val_data["Xflag"], scaler_mri)
        Xmri_te = transform_mri_with_train_params(test_data["Xmri"], test_data["Xflag"], scaler_mri)

        _, _, _, sampler = make_weights_and_sampler(train_data, power=best_params["class_power"])

        train_loader = build_loader(
            Xtr_c, Xmri_tr, train_data["Xflag"], train_data["month_idx"],
            train_data["y_subtype"], train_data["m_subtype"],
            train_data["y_depbin"], train_data["m_depbin"],
            train_data["y_levels"], train_data["m_levels"],
            batch_size=32, sampler=sampler
        )

        val_loader = build_loader(
            Xva_c, Xmri_va, val_data["Xflag"], val_data["month_idx"],
            val_data["y_subtype"], val_data["m_subtype"],
            val_data["y_depbin"], val_data["m_depbin"],
            val_data["y_levels"], val_data["m_levels"],
            batch_size=64, shuffle=False
        )

        test_loader = build_loader(
            Xte_c, Xmri_te, test_data["Xflag"], test_data["month_idx"],
            test_data["y_subtype"], test_data["m_subtype"],
            test_data["y_depbin"], test_data["m_depbin"],
            test_data["y_levels"], test_data["m_levels"],
            batch_size=64, shuffle=False
        )

        model, history, best_val_score = train_one_model(
            train_loader=train_loader,
            val_loader=val_loader,
            params=best_params,
            clinical_input_dim=Xtr_c.shape[-1]
        )

        seed_histories.append(history)

        test_eval = evaluate_model(model, test_loader)
        seed_results.append(test_eval)

        with open(seed_dir / "test_metrics.json", "w", encoding="utf-8") as f:
            json.dump(test_eval, f, indent=2)

        # ------------------------------------------------------------
        # SAVE INDIVIDUAL PREDICTIONS FOR THIS SEED
        # ------------------------------------------------------------
        model.eval()
        rows = []
        with torch.no_grad():
            for i in range(len(test_data["pat"])):
                x_clin_i = torch.tensor(Xte_c[i:i+1], dtype=torch.float32, device=device)
                x_mri_i = torch.tensor(Xmri_te[i:i+1], dtype=torch.float32, device=device)
                x_flag_i = torch.tensor(test_data["Xflag"][i:i+1], dtype=torch.float32, device=device)
                midx_i = torch.tensor(test_data["month_idx"][i:i+1], dtype=torch.long, device=device)

                out_s, out_d, out_l = model(x_clin_i, x_mri_i, x_flag_i, midx_i)

                ps = torch.softmax(out_s, dim=1).cpu().numpy()[0]
                pdp = torch.softmax(out_d, dim=1).cpu().numpy()[0]
                pl = torch.softmax(out_l, dim=1).cpu().numpy()[0]

                rows.append({
                    "seed": seed,
                    "PATNO": int(test_data["pat"][i]),

                    "subtype_true": int(test_data["y_subtype"][i]),
                    "subtype_mask": float(test_data["m_subtype"][i]),
                    "subtype_pred": int(np.argmax(ps)),

                    "depression_true": int(test_data["y_depbin"][i]),
                    "depression_mask": float(test_data["m_depbin"][i]),
                    "depression_pred": int(np.argmax(pdp)),

                    "levels_true": int(test_data["y_levels"][i]),
                    "levels_mask": float(test_data["m_levels"][i]),
                    "levels_pred": int(np.argmax(pl)),

                    "mri_available": float(test_data["Xflag"][i][0]),
                })

        pred_df = pd.DataFrame(rows)
        pred_df.to_csv(seed_dir / "predictions.csv", index=False)
        all_predictions.append(pred_df)

        torch.save(model.state_dict(), seed_dir / "model.pt")

        del model, train_loader, val_loader, test_loader
        gc.collect()

    # ------------------------------------------------------------
    # AGGREGATE METRICS OVER THE 5 SEEDS (mean +/- std, per-class, mean CM)
    # ------------------------------------------------------------
    agg_subtype = aggregate_task_over_seeds(
        [r["subtype"] for r in seed_results], "subtype", N_CLASSES["subtype"], CLASS_NAMES["subtype"]
    )
    agg_depression = aggregate_task_over_seeds(
        [r["depression"] for r in seed_results], "depression", N_CLASSES["depression"], CLASS_NAMES["depression"]
    )
    agg_levels = aggregate_task_over_seeds(
        [r["levels"] for r in seed_results], "levels", N_CLASSES["levels"], CLASS_NAMES["levels"]
    )

    global_scores = [r["global_score"] for r in seed_results]
    global_score_mean = float(np.mean(global_scores))
    global_score_std = float(np.std(global_scores))

    agg_history = aggregate_histories_over_seeds(seed_histories)

    summary = {
        "horizon": h_name,
        "best_search_score": float(study.best_value),

        "subtype": agg_subtype,
        "depression": agg_depression,
        "levels": agg_levels,

        "global_score_mean": global_score_mean,
        "global_score_std": global_score_std,

        "mean_final_train_loss": float(agg_history["mean_train_loss"][-1]),
        "mean_final_val_score": float(agg_history["mean_val_score"][-1]),
    }

    with open(horizon_dir / "summary_metrics.json", "w", encoding="utf-8") as f:
        json.dump(summary, f, indent=2)

    pred_all = pd.concat(all_predictions, ignore_index=True)
    pred_all.to_csv(horizon_dir / f"predictions_{h_name}.csv", index=False)

    print("\nSUMMARY (mean +/- std over {} seeds)".format(N_SEEDS))
    print(json.dumps(summary, indent=2))

    return summary, pred_all, agg_history




# ==========================================================
# 21) RUN ALL HORIZONS (T2 / T4 / T6)
# ==========================================================
all_summaries = []
all_preds = []
agg_histories_by_horizon = {}
cms_by_task_and_horizon = {task: {} for task in ["subtype", "depression", "levels"]}

for h_name, cfg in HORIZONS.items():
    summary, pred_df, agg_history = run_one_horizon(
        h_name=h_name,
        months=cfg["months"],
        last_month=cfg["last_month"]
    )

    all_summaries.append(summary)
    pred_df = pred_df.copy()
    pred_df["time_horizon"] = h_name
    all_preds.append(pred_df)

    agg_histories_by_horizon[h_name] = agg_history

    for task in ["subtype", "depression", "levels"]:
        cms_by_task_and_horizon[task][h_name] = summary[task]["confusion_matrix_mean_rounded"]

preds_df = pd.concat(all_preds, ignore_index=True)
preds_df.to_csv(OUT_DIR / "predictions_all_horizons.csv", index=False)




# ==========================================================
# 22) FIGURE - 3 CONFUSION MATRICES (T2/T4/T6) PER TASK
# ==========================================================
for task in ["subtype", "depression", "levels"]:
    plot_confusion_figure_3horizons(
        task_name=task,
        class_names=CLASS_NAMES[task],
        cms_by_horizon=cms_by_task_and_horizon[task],
        out_path=OUT_DIR / f"global_cm_3horizons_{task}.png"
    )




# ==========================================================
# 23) FIGURE - 3 TRAIN/VAL LOSS CURVES (T2/T4/T6), mean over seeds
# ==========================================================
plot_loss_figure_3horizons(
    agg_histories_by_horizon=agg_histories_by_horizon,
    out_path=OUT_DIR / "loss_curves_3horizons.png"
)




# ==========================================================
# 24) FINAL METRICS TABLE (mean +/- std over the 5 seeds)
# ==========================================================
def fmt_mean_std(m, s, digits=4):
    if m is None or (isinstance(m, float) and np.isnan(m)):
        return "NA"
    return f"{m:.{digits}f} ± {s:.{digits}f}"


final_rows = []
for summary in all_summaries:
    h_name = summary["horizon"]
    for task in ["subtype", "depression", "levels"]:
        agg = summary[task]
        final_rows.append({
            "Horizon": h_name,
            "Task": task,
            "Accuracy": fmt_mean_std(agg["accuracy_mean"], agg["accuracy_std"]),
            "Precision (macro)": fmt_mean_std(agg["precision_macro_mean"], agg["precision_macro_std"]),
            "Recall (macro)": fmt_mean_std(agg["recall_macro_mean"], agg["recall_macro_std"]),
            "F1 (macro)": fmt_mean_std(agg["f1_macro_mean"], agg["f1_macro_std"]),
            "Balanced Accuracy": fmt_mean_std(agg["balanced_accuracy_mean"], agg["balanced_accuracy_std"]),
        })

final_table_df = pd.DataFrame(final_rows)
final_table_df.to_csv(OUT_DIR / "final_metrics_table.csv", index=False)


# ------------------------------------------------------------
# Per-class table (mean +/- std over the 5 seeds)
# ------------------------------------------------------------
per_class_rows = []
for summary in all_summaries:
    h_name = summary["horizon"]
    for task in ["subtype", "depression", "levels"]:
        agg = summary[task]
        for cname, cstats in agg["per_class"].items():
            per_class_rows.append({
                "Horizon": h_name,
                "Task": task,
                "Class": cname,
                "Precision": fmt_mean_std(cstats["precision_mean"], cstats["precision_std"]),
                "Recall": fmt_mean_std(cstats["recall_mean"], cstats["recall_std"]),
                "F1": fmt_mean_std(cstats["f1_mean"], cstats["f1_std"]),
            })

per_class_table_df = pd.DataFrame(per_class_rows)
per_class_table_df.to_csv(OUT_DIR / "final_per_class_table.csv", index=False)


# ------------------------------------------------------------
# Loss summary table (mean Train Loss / mean Val Loss over the 5 seeds)
# ------------------------------------------------------------
loss_summary_rows = []
for summary in all_summaries:
    loss_summary_rows.append({
        "Horizon": summary["horizon"],
        "Mean Final Train Loss (5 seeds)": round(summary["mean_final_train_loss"], 4),
        "Mean Final Val Score (5 seeds)": round(summary["mean_final_val_score"], 4),
        "Global Score": fmt_mean_std(summary["global_score_mean"], summary["global_score_std"]),
    })

loss_summary_df = pd.DataFrame(loss_summary_rows)
loss_summary_df.to_csv(OUT_DIR / "loss_summary_table.csv", index=False)




# ==========================================================
# 25) AUTOMATIC DISPLAY AT THE END
# ==========================================================
print("\n" + "=" * 80)
print("FINAL METRICS TABLE (Mean +/- Std over {} seeds, no folds)".format(N_SEEDS))
print("=" * 80)
display(final_table_df)

print("\n" + "=" * 80)
print("PER-CLASS METRICS TABLE (Mean +/- Std over {} seeds)".format(N_SEEDS))
print("=" * 80)
display(per_class_table_df)

print("\n" + "=" * 80)
print("MEAN TRAIN LOSS / MEAN VALIDATION SCORE OVER {} SEEDS".format(N_SEEDS))
print("=" * 80)
display(loss_summary_df)

print("\nSaved outputs in:", OUT_DIR)
print(" -", OUT_DIR / "final_metrics_table.csv")
print(" -", OUT_DIR / "final_per_class_table.csv")
print(" -", OUT_DIR / "loss_summary_table.csv")
print(" -", OUT_DIR / "predictions_all_horizons.csv")
print(" - global_cm_3horizons_{subtype,depression,levels}.png")
print(" - loss_curves_3horizons.png")
```
